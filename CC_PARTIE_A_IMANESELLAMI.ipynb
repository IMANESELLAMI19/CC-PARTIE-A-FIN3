## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
- **Nom** : Régression linéaire (moindres carrés ordinaires) – *Linear Regression (OLS)*. **Famille** : régression supervisée ; objectif : prédire une variable cible quantitative.
- **Intuition** : on cherche l'hyperplan qui passe au plus près de l'ensemble des observations.
- **Étapes** : données d'entrée (X, y) → calcul des coefficients qui minimisent l'erreur quadratique → prédiction ŷ = Xβ + b pour de nouvelles observations.
- **Critère** : $\min_{\beta,\,b}\ \sum_{i=1}^{n}\left(y_i-\beta^\top x_i-b\right)^2$
- **Symboles** : $n$ = nombre d'observations ; $x_i$ = vecteur des variables explicatives de l'observation $i$ ; $y_i$ = valeur observée ; $\beta$ = coefficients ; $b$ = ordonnée à l'origine.
- **Paramètres** (appris) : $\beta$ et $b$. **Hyperparamètres** : pratiquement aucun (seulement des options comme `fit_intercept`).
- **Avantage 1** : modèle très interprétable (un coefficient = un effet marginal).
- **Avantage 2** : rapide à entraîner, solution unique et exacte.
- **Limite 1** : sensible à la multicolinéarité et aux valeurs aberrantes.
- **Limite 2** : ne capte que des relations linéaires entre variables et cible.

Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
x = rng.uniform(0, 10, 40)
y = 2*x + 3 + rng.normal(0, 2, 40)
b1, b0 = np.polyfit(x, y, 1)

plt.figure(figsize=(6, 4))
plt.scatter(x, y, label="Observations (données synthétiques)")
xs = np.linspace(0, 10, 100)
plt.plot(xs, b1*xs + b0, color="red", label="Droite ajustée ŷ = βx + b")
for xi, yi in list(zip(x, y))[:8]:
    plt.plot([xi, xi], [yi, b1*xi + b0], color="gray", ls="--", lw=0.8)
plt.plot([], [], color="gray", ls="--", label="Résidus (erreurs)")
plt.xlabel("Variable explicative x"); plt.ylabel("Cible y")
plt.title("Régression linéaire : minimiser la somme des carrés des résidus")
plt.legend(); plt.show()
```

**Lecture du schéma** : chaque point bleu est une observation ; la droite rouge est le modèle appris ; les segments gris pointillés sont les résidus. L'algorithme choisit la droite qui rend la somme des carrés de ces segments la plus petite possible.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion)** : estimer le chiffre d'affaires mensuel d'un magasin à partir du budget marketing, du prix et de la saisonnalité.
- **Cas d'usage 2** : prédire le prix d'un bien immobilier à partir de sa surface, de son âge et de sa localisation.
- **Données et prérequis** : variables numériques (ou catégorielles encodées), cible continue, absence de valeurs manquantes, relation approximativement linéaire, variables peu corrélées entre elles.
- **Cas peu approprié** : prédire une probabilité de défaut de paiement (cible binaire) ou une relation fortement non linéaire (ex. : courbe de saturation de la demande).
- **Métriques adaptées** : MAE, RMSE, R², évaluées sur des données de test.

### Analyse de l'expérience Python
- **Entrées** : `Xr_ent`, `yr_ent` (données d'entraînement) pour l'ajustement ; `Xr_test`, `yr_test` (données de test) pour l'évaluation. **Hyperparamètres** : aucun, valeurs par défaut de `LinearRegression()`.
- **Sortie** : MAE = 9,3535, soit l'écart absolu moyen entre les valeurs prédites et les valeurs réelles, exprimé dans l'unité de la cible.
- **Ce que cela permet** : disposer d'une valeur de référence (*baseline*) pour comparer les modèles suivants.
- **Ce que cela ne permet pas** : juger si 9,35 est « bon » sans connaître l'échelle de la cible, ni connaître la répartition des erreurs (une MAE ne distingue pas les grosses erreurs rares des petites erreurs fréquentes), ni vérifier les hypothèses linéaires.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
- **Nom** : Régression Ridge (régularisation L2) – *Ridge Regression*. **Famille** : régression supervisée ; objectif : prédire une cible continue en limitant le surapprentissage.
- **Intuition** : c'est une régression linéaire à laquelle on ajoute une pénalité qui empêche les coefficients de devenir trop grands.
- **Étapes** : standardisation des variables → minimisation de l'erreur quadratique + pénalité → prédiction ŷ = Xβ + b.
- **Critère** : $\min_{\beta,\,b}\ \sum_{i=1}^{n}\left(y_i-\beta^\top x_i-b\right)^2+\alpha\sum_{j=1}^{p}\beta_j^2$
- **Symboles** : $n$ observations, $p$ variables ; $y_i$ = cible ; $x_i$ = variables ; $\beta_j$ = coefficient de la variable $j$ ; $\alpha \ge 0$ = force de la pénalité.
- **Paramètres** (appris) : $\beta$ et $b$. **Hyperparamètre** : $\alpha$ (ici 1.0), à choisir par validation croisée.
- **Avantage 1** : stabilise les coefficients en cas de multicolinéarité.
- **Avantage 2** : réduit la variance et le risque de surapprentissage.
- **Limite 1** : ne met aucun coefficient exactement à zéro (pas de sélection de variables).
- **Limite 2** : exige une standardisation et le choix de $\alpha$ ; les coefficients sont biaisés, donc moins directement interprétables.

### Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
from sklearn.datasets import make_regression

X, y = make_regression(n_samples=100, n_features=6, noise=10, random_state=0)
alphas = np.logspace(-2, 4, 50)
coefs = [Ridge(alpha=a).fit(X, y).coef_ for a in alphas]

plt.figure(figsize=(6, 4))
plt.plot(alphas, coefs)
plt.xscale("log")
plt.xlabel("α (échelle logarithmique)"); plt.ylabel("Valeur des coefficients β")
plt.title("Ridge : les coefficients se contractent quand α augmente\n(données synthétiques, 1 courbe = 1 variable)")
plt.show()
```

**Lecture du schéma** : chaque courbe représente le coefficient d'une variable. À gauche (α faible), on retrouve les coefficients de la régression linéaire ; en allant vers la droite (α élevé), ils se rapprochent tous de zéro sans jamais l'atteindre.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/marketing)** : prévoir les ventes à partir de nombreux indicateurs fortement corrélés (dépenses publicitaires par canal, promotions, trafic web).
- **Cas d'usage 2** : modéliser le rendement d'un portefeuille à partir de facteurs économiques corrélés entre eux.
- **Données et prérequis** : variables numériques standardisées, cible continue, nombre de variables élevé ou variables corrélées.
- **Cas peu approprié** : quand on veut éliminer automatiquement les variables inutiles (Lasso est alors préférable), ou en présence de relations non linéaires marquées.
- **Métriques adaptées** : MAE, RMSE, R², avec choix de $\alpha$ par validation croisée sur l'ensemble d'entraînement.

### Analyse de l'expérience Python
- **Entrées** : mêmes jeux d'entraînement et de test que précédemment. **Hyperparamètres** : `alpha=1.0` (valeur par défaut, non optimisée) ; `StandardScaler` intégré dans un pipeline pour standardiser les variables avant l'ajustement.
- **Sortie** : MAE = 9,3678, à comparer avec 9,3535 pour la régression linéaire.
- **Ce que cela permet** : constater que, avec $\alpha = 1$, la pénalisation n'améliore pas la MAE sur ce jeu de test ; l'écart (environ 0,014) est très faible.
- **Ce que cela ne permet pas** : conclure que Ridge est moins bon, car l'écart est minime et obtenu sur une seule découpe des données. On ne peut pas non plus dire qu'un autre $\alpha$ ne ferait pas mieux : il faudrait le tester par validation croisée, sans utiliser le jeu de test pour le réglage.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Présentation
- **Nom** : Régression Lasso (régularisation L1) – *Lasso Regression (Least Absolute Shrinkage and Selection Operator)*. **Famille** : régression supervisée ; objectif : prédire une cible continue tout en sélectionnant les variables utiles.
- **Intuition** : c'est une régression linéaire avec une pénalité sur la somme des valeurs absolues des coefficients, ce qui force les coefficients peu utiles à devenir exactement nuls.
- **Étapes** : standardisation des variables → minimisation de l'erreur quadratique + pénalité L1 (par descente de coordonnées) → prédiction ŷ = Xβ + b avec un sous-ensemble de variables.
- **Critère** : $\min_{\beta,\,b}\ \frac{1}{2n}\sum_{i=1}^{n}\left(y_i-\beta^\top x_i-b\right)^2+\alpha\sum_{j=1}^{p}|\beta_j|$
- **Symboles** : $n$ observations, $p$ variables ; $y_i$ = cible ; $x_i$ = variables ; $\beta_j$ = coefficient de la variable $j$ ; $b$ = ordonnée à l'origine ; $\alpha \ge 0$ = force de la pénalité.
- **Paramètres** (appris) : $\beta$ et $b$. **Hyperparamètres** : $\alpha$ (ici 0,1) et `max_iter` (ici 10 000, limite d'itérations de l'optimiseur).
- **Avantage 1** : effectue une sélection automatique de variables (modèle parcimonieux).
- **Avantage 2** : réduit le surapprentissage quand il y a beaucoup de variables.
- **Limite 1** : instable en cas de variables fortement corrélées (il en retient une presque arbitrairement).
- **Limite 2** : coefficients biaisés ; sensible au choix de $\alpha$ et à la standardisation.

### Schéma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.preprocessing import StandardScaler

# Données synthétiques : 8 variables, dont seulement 3 sont réellement informatives
X, y = make_regression(n_samples=150, n_features=8, n_informative=3,
                       noise=5, random_state=0)
X = StandardScaler().fit_transform(X)

alphas = np.logspace(-2, 1.5, 40)
coefs = [Lasso(alpha=a, max_iter=10000).fit(X, y).coef_ for a in alphas]

plt.figure(figsize=(6, 4))
plt.plot(alphas, coefs)
plt.xscale("log")
plt.axhline(0, color="black", lw=0.8)
plt.xlabel("α (échelle logarithmique)")
plt.ylabel("Valeur des coefficients β")
plt.title("Lasso : les coefficients atteignent exactement 0\n(données synthétiques, 1 courbe = 1 variable)")
plt.show()
```

**Lecture du schéma** : chaque courbe est le coefficient d'une variable. À gauche (α faible), presque toutes les variables ont un coefficient non nul. Quand α augmente, les courbes touchent la ligne horizontale à 0 les unes après les autres : ces variables sont éliminées du modèle. Les dernières à rester sont les plus utiles pour prédire la cible.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/commerce)** : identifier, parmi des dizaines d'indicateurs (canaux publicitaires, prix, promotions, météo), ceux qui expliquent réellement les ventes.
- **Cas d'usage 2 (finance)** : construire un modèle de scoring ou de prévision de rendement parcimonieux à partir de nombreux ratios financiers.
- **Données et prérequis** : variables numériques standardisées, cible continue, nombre de variables élevé (parfois supérieur au nombre d'observations).
- **Cas peu approprié** : variables très corrélées entre elles et toutes utiles (Ridge ou Elastic Net sont plus stables), ou relation fortement non linéaire.
- **Métriques adaptées** : MAE, RMSE, R², plus le nombre de coefficients non nuls ; $\alpha$ choisi par validation croisée.

### Analyse de l'expérience Python
- **Entrées** : `Xr_ent`, `yr_ent` pour l'entraînement ; `Xr_test`, `yr_test` pour l'évaluation. **Hyperparamètres** : `alpha=0.1`, `max_iter=10000` ; `StandardScaler` dans un pipeline pour standardiser les variables.
- **Sortie** : R² = 0,9860, c'est-à-dire qu'environ 98,6 % de la variance de la cible sur le jeu de test est expliquée par le modèle.
- **Ce que cela permet** : constater que le modèle reproduit très bien la cible sur des données non vues pendant l'entraînement.
- **Ce que cela ne permet pas** : comparer directement ce résultat aux MAE des modèles précédents (métrique différente) ; connaître l'erreur en unités de la cible ; savoir quelles variables ont été éliminées (les coefficients ne sont pas affichés) ; confirmer que $\alpha = 0{,}1$ est optimal (valeur fixée, non validée) ; généraliser à partir d'une seule découpe train/test.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
- **Nom** : Régression polynomiale (ici régularisée par Ridge) – *Polynomial Regression*. **Famille** : régression supervisée ; objectif : modéliser une relation non linéaire entre variables et cible continue.
- **Intuition** : on ajoute aux variables d'origine leurs puissances et leurs produits, puis on ajuste un modèle linéaire sur ces nouvelles variables.
- **Étapes** : standardisation → création des variables polynomiales (degré 2) → régression Ridge sur ces variables → prédiction.
- **Critère** : $\hat{y}=b+\sum_{j}\beta_j x_j+\sum_{j\le k}\beta_{jk}\,x_j x_k$, avec $\min \sum_{i=1}^{n}(y_i-\hat{y}_i)^2+\alpha\|\beta\|_2^2$
- **Symboles** : $x_j$ = variable $j$ ; $x_j x_k$ = terme d'interaction (ou carré si $j=k$) ; $\beta_j,\beta_{jk}$ = coefficients ; $b$ = ordonnée à l'origine ; $\alpha$ = force de la pénalité.
- **Paramètres** (appris) : $b$, $\beta_j$ et $\beta_{jk}$. **Hyperparamètres** : degré (ici 2), $\alpha$ (ici 10), `include_bias=False`.
- **Avantage 1** : capte des courbures et des interactions entre variables tout en restant un modèle linéaire dans ses coefficients.
- **Avantage 2** : simple à mettre en œuvre et à interpréter pour un faible degré.
- **Limite 1** : le nombre de variables explose avec le degré et le nombre de variables (risque de surapprentissage).
- **Limite 2** : extrapole mal en dehors de la plage des données d'entraînement.

Schèma explicatif

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

rng = np.random.default_rng(1)
x = np.sort(rng.uniform(-3, 3, 60))
y = 0.8*x**2 - x + rng.normal(0, 1.2, 60)   # relation courbe (données synthétiques)
X = x.reshape(-1, 1)

droite = LinearRegression().fit(X, y)
poly2 = make_pipeline(PolynomialFeatures(degree=2, include_bias=False),
                      LinearRegression()).fit(X, y)
xs = np.linspace(-3, 3, 200).reshape(-1, 1)

plt.figure(figsize=(6, 4))
plt.scatter(x, y, label="Observations (données synthétiques)")
plt.plot(xs, droite.predict(xs), color="gray", ls="--", label="Régression linéaire (degré 1)")
plt.plot(xs, poly2.predict(xs), color="red", label="Régression polynomiale (degré 2)")
plt.xlabel("Variable explicative x"); plt.ylabel("Cible y")
plt.title("Régression polynomiale : ajuster une courbe avec un modèle linéaire")
plt.legend(); plt.show()
```

**Lecture du schéma** : les points forment une courbe en « U ». La droite grise en pointillés (degré 1) ne peut pas la suivre et laisse de grandes erreurs. La courbe rouge (degré 2), obtenue en ajoutant le terme $x^2$, épouse la forme des données.

### Contextes d'utilisation

- **Cas peu approprié** : beaucoup de variables avec un degré élevé (explosion du nombre de termes et surapprentissage), ou prédiction hors de la plage observée.
- **Métriques adaptées** : MAE, RMSE, R² sur jeu de test ; degré et $\alpha$ choisis par validation croisée.

### Présentation
- **Nom** : Arbre de régression (CART) – *Regression Tree (Decision Tree Regressor)*. **Famille** : régression supervisée ; objectif : prédire une cible continue par une suite de règles « si… alors… ».
- **Intuition** : on découpe l'espace des variables en zones rectangulaires, et chaque zone prédit la moyenne des cibles qu'elle contient.
- **Étapes** : données (X, y) → recherche récursive du meilleur seuil de coupure sur une variable → arrêt à la profondeur maximale → prédiction = moyenne de la feuille atteinte.
- **Critère** : à chaque nœud, on choisit la variable $j$ et le seuil $s$ qui minimisent $\frac{n_G}{n}\,\mathrm{MSE}_G+\frac{n_D}{n}\,\mathrm{MSE}_D$.
- **Symboles** : $n$ = observations du nœud ; $n_G, n_D$ = observations envoyées à gauche (variable $j \le s$) et à droite ; $\mathrm{MSE}_G, \mathrm{MSE}_D$ = variance de la cible dans chaque groupe.
- **Paramètres** (appris) : variables et seuils de coupure, valeurs des feuilles. **Hyperparamètres** : `max_depth` (ici 5), `random_state` (ici `GRAINE`, pour la reproductibilité).
- **Avantage 1** : très lisible (règles visualisables) et sans standardisation nécessaire.
- **Avantage 2** : capte des relations non linéaires et des interactions automatiquement.
- **Limite 1** : instable (un petit changement des données peut modifier l'arbre) et sujet au surapprentissage si trop profond.
- **Limite 2** : prédictions constantes par morceaux, incapables d'extrapoler hors de la plage d'entraînement.

Schème explicatif

Chaque palier horizontal est une feuille de l'arbre : tous les points tombant dans cette zone reçoivent la même prédiction (la moyenne de leurs cibles). - **Cas peu approprié** : relation simple et régulière (quasi linéaire), où un arbre produit des paliers au lieu d'une droite ; ou prévision hors de la plage des données (tendance temporelle à prolonger).
- **Métriques adaptées** : MAE, RMSE, R² sur jeu de test ; profondeur choisie par validation croisée.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
- **Nom** : Forêt aléatoire pour la régression – *Random Forest Regressor*. **Famille** : régression supervisée (méthode d'ensemble) ; objectif : prédire une cible continue en moyennant de nombreux arbres.
- **Intuition** : un seul arbre est instable ; en moyennant plusieurs arbres entraînés sur des échantillons différents, les erreurs individuelles se compensent.
- **Étapes** : tirage de $B$ échantillons bootstrap (avec remise) → entraînement d'un arbre par échantillon → prédiction = moyenne des prédictions des $B$ arbres.
- **Critère** : $\hat{y}(x)=\frac{1}{B}\sum_{b=1}^{B}T_b(x)$
- **Symboles** : $B$ = nombre d'arbres (`n_estimators`) ; $T_b(x)$ = prédiction du $b$-ième arbre pour l'observation $x$ ; chaque arbre est construit sur son propre échantillon bootstrap.
- **Paramètres** (appris) : seuils et feuilles de chaque arbre. **Hyperparamètres** : `n_estimators` (ici 80), `random_state` (ici `GRAINE`) ; autres, comme `max_depth` ou `max_features`, à leurs valeurs par défaut.
- **Avantage 1** : bien plus stable et moins sujet au surapprentissage qu'un arbre isolé.
- **Avantage 2** : peu de réglages nécessaires, gère les non-linéarités et les interactions.
- **Limite 1** : moins interprétable (80 arbres au lieu d'un) et plus coûteuse en calcul et en mémoire.
- **Limite 2** : ne peut pas extrapoler en dehors de la plage des cibles vues à l'entraînement.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

rng = np.random.default_rng(0)
x = np.sort(rng.uniform(0, 10, 80))
y = np.sin(x) * 3 + rng.normal(0, 0.8, 80)   # données synthétiques
X = x.reshape(-1, 1)
xs = np.linspace(0, 10, 500).reshape(-1, 1)

plt.figure(figsize=(6, 4))
plt.scatter(x, y, s=15, color="black", label="Observations (données synthétiques)")
preds = []
for b in range(30):
    idx = rng.integers(0, len(x), len(x))          # échantillon bootstrap
    arbre = DecisionTreeRegressor(max_depth=5, random_state=b).fit(X[idx], y[idx])
    p = arbre.predict(xs)
    preds.append(p)
    plt.plot(xs, p, color="lightblue", lw=0.8, alpha=0.7)
plt.plot([], [], color="lightblue", label="Arbres individuels (bootstrap)")
plt.plot(xs, np.mean(preds, axis=0), color="red", lw=2, label="Moyenne de la forêt")
plt.xlabel("Variable explicative x"); plt.ylabel("Cible y")
plt.title("Forêt aléatoire : agrégation (moyenne) de plusieurs arbres")
plt.legend(); plt.show()
```

**Lecture du schéma** : chaque courbe bleu clair est la prédiction d'un arbre entraîné sur un échantillon bootstrap différent ; elles varient beaucoup d'un arbre à l'autre. La courbe rouge épaisse est la moyenne de ces arbres : elle est plus lisse et plus stable que chacun d'eux.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion)** : prévoir le montant d'un sinistre, la valeur d'un bien ou la demande d'un produit à partir de nombreuses variables hétérogènes.
- **Cas d'usage 2 (logistique/commerce international)** : estimer les délais ou les coûts de transport selon l'itinéraire, le poids, la saison et le transporteur.
- **Données et prérequis** : tableau de variables numériques ou catégorielles encodées, cible continue, effectif moyen à grand ; pas de standardisation nécessaire.
- **Cas peu approprié** : extrapolation d'une tendance (série temporelle croissante), ou contexte exigeant des règles simples et transparentes (par exemple pour justifier une décision à un régulateur).
- **Métriques adaptées** : MAE, RMSE, R² sur jeu de test ; erreur *out-of-bag* ou validation croisée pour le réglage.

### Analyse de l'expérience Python
- **Entrées** : `Xr_ent`, `yr_ent` pour l'entraînement ; `Xr_test`, `yr_test` pour l'évaluation. **Hyperparamètres** : `n_estimators=80`, `random_state=GRAINE` ; les autres gardent leurs valeurs par défaut.
- **Sortie** : MAE = 24,3738, dans l'unité de la cible.
- **Ce que cela permet** : constater que la forêt (24,37) fait nettement mieux que l'arbre seul de profondeur 5 (37,72), soit une baisse d'environ 35 % de la MAE, ce qui est cohérent avec l'effet stabilisateur de la moyenne.
- **Ce que cela ne permet pas** : conclure qu'elle est meilleure que les modèles linéaires : si les données sont les mêmes, leur MAE (environ 9,35 à 10,90) reste bien plus faible ; expliquer cet écart sans analyse supplémentaire ; savoir si un autre nombre d'arbres ou une autre profondeur donnerait mieux (non testé) ; généraliser à partir d'une seule découpe train/test.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
- **Nom** : Gradient Boosting pour la régression (ici version par histogrammes) – *Histogram-based Gradient Boosting Regressor*. **Famille** : régression supervisée (méthode d'ensemble) ; objectif : prédire une cible continue en corrigeant progressivement les erreurs.
- **Intuition** : on construit les arbres un par un, chacun apprenant à corriger les erreurs (résidus) laissées par l'ensemble des arbres précédents.
- **Étapes** : prédiction initiale (la moyenne de la cible) → calcul des résidus → ajustement d'un petit arbre sur ces résidus → ajout de sa prédiction, pondérée, au modèle → répétition `max_iter` fois. La version « histogramme » regroupe les valeurs des variables en classes pour accélérer le calcul.
- **Critère** : $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$, où $h_m$ minimise l'erreur sur les résidus $r_i=y_i-F_{m-1}(x_i)$.
- **Symboles** : $F_m$ = modèle après $m$ arbres ; $h_m$ = $m$-ième arbre ; $\eta$ = taux d'apprentissage (`learning_rate`) ; $r_i$ = résidu de l'observation $i$.
- **Paramètres** (appris) : structure et feuilles de chaque arbre. **Hyperparamètres** : `max_iter` (ici 80, nombre d'arbres), `random_state` (ici `GRAINE`) ; `learning_rate`, `max_depth`, etc. à leurs valeurs par défaut.
- **Avantage 1** : souvent très précis sur des données tabulaires, avec capture des non-linéarités et des interactions.
- **Avantage 2** : la version par histogrammes est rapide sur de grands jeux de données.
- **Limite 1** : sensible au réglage (nombre d'arbres, taux d'apprentissage) et risque de surapprentissage.
- **Limite 2** : peu interprétable et ne peut pas extrapoler hors de la plage d'entraînement.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

rng = np.random.default_rng(0)
x = np.sort(rng.uniform(0, 10, 80))
y = np.sin(x) * 3 + rng.normal(0, 0.4, 80)   # données synthétiques
X = x.reshape(-1, 1)
xs = np.linspace(0, 10, 500).reshape(-1, 1)

eta = 0.5
pred_x = np.full(len(x), y.mean())
pred_xs = np.full(len(xs), y.mean())
etapes = [0, 1, 4, 20]
fig, axes = plt.subplots(1, 4, figsize=(14, 3), sharey=True)
k = 0
for m in range(max(etapes) + 1):
    if m in etapes:
        ax = axes[k]; k += 1
        ax.scatter(x, y, s=8, color="gray")
        ax.plot(xs, pred_xs, color="red")
        ax.set_title(f"{m} arbre(s)")
        ax.set_xlabel("x")
    arbre = DecisionTreeRegressor(max_depth=2).fit(X, y - pred_x)   # arbre sur les résidus
    pred_x = pred_x + eta * arbre.predict(X)
    pred_xs = pred_xs + eta * arbre.predict(xs)
axes[0].set_ylabel("Cible y")
fig.suptitle("Gradient Boosting : chaque arbre corrige les résidus des précédents (données synthétiques)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : de gauche à droite, on ajoute des arbres. Les points gris sont les observations ; la courbe rouge est la prédiction du modèle. Sans arbre, la prédiction est une droite horizontale (la moyenne). À chaque nouvel arbre, la courbe se rapproche un peu plus des points, car l'arbre ajouté corrige les écarts restants.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion)** : prévoir la valeur d'un client (*customer lifetime value*) ou le montant d'un sinistre à partir d'un grand nombre de variables hétérogènes.
- **Cas d'usage 2 (logistique/commerce)** : prévoir la demande d'un produit ou le délai de livraison à partir de variables de prix, de saison, de lieu et d'historique.
- **Données et prérequis** : données tabulaires numériques (la version par histogrammes accepte aussi les valeurs manquantes), cible continue, effectif moyen à grand ; pas de standardisation nécessaire.
- **Cas peu approprié** : très petit jeu de données (surapprentissage), ou contexte exigeant un modèle simple et transparent à expliquer.
- **Métriques adaptées** : MAE, RMSE, R² sur jeu de test ; réglage par validation croisée ou arrêt précoce sur un jeu de validation.

### Analyse de l'expérience Python
- **Entrées** : `Xr_ent`, `yr_ent` pour l'entraînement ; `Xr_test`, `yr_test` pour l'évaluation. **Hyperparamètres** : `max_iter=80`, `random_state=GRAINE` ; les autres gardent leurs valeurs par défaut.
- **Sortie** : MAE = 23,2661, dans l'unité de la cible.
- **Ce que cela permet** : si les données sont les mêmes que pour les modèles 1 à 6, constater que cette MAE est légèrement meilleure que celle de la forêt aléatoire (24,3738) et bien meilleure que celle de l'arbre seul (37,7171).
- **Ce que cela ne permet pas** : conclure qu'il est meilleur que les modèles linéaires, dont la MAE (environ 9,35 à 10,90) reste plus faible ; affirmer que l'écart avec la forêt (environ 1,1) est significatif, car il vient d'une seule découpe train/test ; savoir si un autre nombre d'itérations ou taux d'apprentissage donnerait mieux (non testé).

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
- **Nom** : Régression logistique – *Logistic Regression*. **Famille** : classification supervisée ; objectif : estimer la probabilité d'appartenance à une classe (ici binaire) et en déduire une décision.
- **Intuition** : on calcule un score linéaire des variables, puis on le transforme en probabilité entre 0 et 1 grâce à la fonction sigmoïde.
- **Étapes** : standardisation → apprentissage des coefficients par minimisation de la perte logistique (avec pénalité L2 par défaut) → probabilité $P(y=1\mid x)$ → classe 1 si cette probabilité dépasse 0,5.
- **Critère** : $P(y=1\mid x)=\dfrac{1}{1+e^{-(\beta^\top x+b)}}$, avec minimisation de $-\sum_{i}\left[y_i\ln p_i+(1-y_i)\ln(1-p_i)\right]$ plus une pénalité L2.
- **Symboles** : $x$ = variables ; $\beta$ = coefficients ; $b$ = ordonnée à l'origine ; $p_i=P(y_i=1\mid x_i)$ ; $y_i\in\{0,1\}$ = classe observée.
- **Paramètres** (appris) : $\beta$ et $b$. **Hyperparamètres** : `max_iter` (ici 1000, limite d'itérations de l'optimiseur), `random_state` (ici `GRAINE`) ; force de régularisation `C` et seuil de décision à leurs valeurs par défaut.
- **Avantage 1** : fournit des probabilités et des coefficients interprétables (effet de chaque variable sur la probabilité).
- **Avantage 2** : rapide, robuste, bon modèle de référence en classification.
- **Limite 1** : frontière de décision linéaire (ne capte pas les interactions ni les non-linéarités sans variables supplémentaires).
- **Limite 2** : sensible aux variables corrélées et aux déséquilibres de classes si on garde le seuil de 0,5.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

rng = np.random.default_rng(0)
x = np.concatenate([rng.normal(-1.5, 1.2, 50), rng.normal(1.5, 1.2, 50)])
y = np.array([0]*50 + [1]*50)                  # données synthétiques
modele = LogisticRegression().fit(x.reshape(-1, 1), y)
xs = np.linspace(-6, 6, 300).reshape(-1, 1)
proba = modele.predict_proba(xs)[:, 1]

plt.figure(figsize=(6, 4))
plt.scatter(x[y == 0], y[y == 0], color="tab:blue", label="Classe 0")
plt.scatter(x[y == 1], y[y == 1], color="tab:orange", label="Classe 1")
plt.plot(xs, proba, color="red", label="P(y=1 | x) (sigmoïde)")
plt.axhline(0.5, color="gray", ls="--", label="Seuil 0,5")
plt.xlabel("Variable explicative x"); plt.ylabel("Probabilité / classe")
plt.title("Régression logistique : de la variable à la probabilité (données synthétiques)")
plt.legend(); plt.show()
```

**Lecture du schéma** : les points bleus (classe 0) sont en bas, les points orange (classe 1) en haut. La courbe rouge en S donne la probabilité d'appartenir à la classe 1 selon la valeur de x. Là où elle croise la ligne grise pointillée (0,5), le modèle bascule de la classe 0 à la classe 1 : c'est la frontière de décision.

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : estimer la probabilité de défaut de paiement d'un emprunteur (scoring de crédit) à partir de ses revenus, de son endettement et de son historique.
- **Cas d'usage 2 (marketing)** : prédire si un client va se désabonner (*churn*) ou répondre à une campagne.
- **Données et prérequis** : variables numériques standardisées (ou catégorielles encodées), cible binaire (ou multiclasse), effectif suffisant dans chaque classe.
- **Cas peu approprié** : frontière entre classes très non linéaire (par exemple une classe au centre et l'autre autour), sauf à ajouter des variables transformées.
- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC ; accuracy à éviter si les classes sont déséquilibrées.

### Analyse de l'expérience Python
- **Entrées** : `Xc_ent`, `yc_ent` pour l'entraînement ; `Xc_test`, `yc_test` pour l'évaluation (jeu de classification, différent de celui des régressions). **Hyperparamètres** : `max_iter=1000`, `random_state=GRAINE` ; `StandardScaler` dans un pipeline ; les autres gardent leurs valeurs par défaut.
- **Sortie** : F1 = 0,8113, moyenne harmonique de la précision et du rappel pour la classe positive (valeur comprise entre 0 et 1, 1 étant le meilleur).
- **Ce que cela permet** : constater un bon équilibre global entre précision et rappel sur la classe positive, et disposer d'une valeur de référence pour comparer les classifieurs suivants.
- **Ce que cela ne permet pas** : connaître séparément la précision et le rappel, ni la matrice de confusion ; juger la performance sur la classe négative (le F1 ne l'évalue pas) ; savoir si le seuil de 0,5 ou `C` sont adaptés (valeurs par défaut, non validées) ; généraliser à partir d'une seule découpe train/test.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
- **Nom** : K plus proches voisins – *K-Nearest Neighbors (KNN)*. **Famille** : classification supervisée (méthode à base d'instances) ; objectif : prédire la classe d'une observation d'après celle de ses voisins.
- **Intuition** : « dis-moi qui sont tes voisins, je te dirai qui tu es » : une observation reçoit la classe majoritaire parmi les $k$ points d'entraînement les plus proches.
- **Étapes** : standardisation → stockage des données d'entraînement (aucun calcul de coefficients) → pour une nouvelle observation, calcul des distances à tous les points d'entraînement → vote majoritaire des $k$ plus proches.
- **Critère** : $\hat{y}(x)=\underset{c}{\arg\max}\sum_{i\in V_k(x)}\mathbb{1}\left[y_i=c\right]$, avec la distance euclidienne $d(x,x_i)=\sqrt{\sum_{j=1}^{p}(x_j-x_{ij})^2}$.
- **Symboles** : $V_k(x)$ = ensemble des $k$ voisins les plus proches de $x$ ; $y_i$ = classe du voisin $i$ ; $c$ = classe candidate ; $p$ = nombre de variables ; $\mathbb{1}[\cdot]$ = 1 si la condition est vraie, 0 sinon.
- **Paramètres** : aucun paramètre appris (le modèle mémorise les données). **Hyperparamètres** : `n_neighbors` ($k$, ici 5), métrique de distance et pondération (valeurs par défaut).
- **Avantage 1** : très simple, sans phase d'apprentissage, capable de frontières non linéaires.
- **Avantage 2** : peu d'hypothèses sur la forme des données.
- **Limite 1** : sensible à l'échelle des variables (d'où la standardisation) et à la dimension élevée ; prédiction lente sur de grands jeux de données.
- **Limite 2** : sensible au choix de $k$ et aux classes déséquilibrées.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)
A = rng.normal([-1, -1], 0.9, (30, 2))     # classe 0 (données synthétiques)
B = rng.normal([1.5, 1.5], 0.9, (30, 2))   # classe 1
X = np.vstack([A, B]); y = np.array([0]*30 + [1]*30)
nouveau = np.array([0.4, 0.4]); k = 5

d = np.linalg.norm(X - nouveau, axis=1)
voisins = np.argsort(d)[:k]

plt.figure(figsize=(6, 5))
plt.scatter(*A.T, color="tab:blue", label="Classe 0")
plt.scatter(*B.T, color="tab:orange", label="Classe 1")
for i in voisins:
    plt.plot([nouveau[0], X[i, 0]], [nouveau[1], X[i, 1]], color="gray", ls="--", lw=0.9)
plt.scatter(*X[voisins].T, s=140, facecolors="none", edgecolors="black", label=f"{k} plus proches voisins")
plt.scatter(*nouveau, color="red", marker="*", s=250, label="Nouvelle observation")
plt.xlabel("Variable 1"); plt.ylabel("Variable 2")
plt.title(f"KNN (k={k}) : vote des voisins les plus proches (données synthétiques)")
plt.legend(); plt.show()
```

**Lecture du schéma** : les points bleus et orange sont les observations connues de chaque classe. L'étoile rouge est la nouvelle observation à classer. Les traits pointillés la relient à ses 5 plus proches voisins (entourés en noir) : la classe la plus représentée parmi eux devient sa prédiction.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/commerce)** : classer un nouveau client dans un segment (acheteur ou non) d'après les clients qui lui ressemblent le plus.
- **Cas d'usage 2 (finance)** : repérer des transactions ou dossiers de crédit semblables à des cas déjà étiquetés (fraude ou non) pour aider à la décision.
- **Données et prérequis** : variables numériques standardisées, peu de variables, effectif modéré ; nécessite de conserver tout le jeu d'entraînement en mémoire.
- **Cas peu approprié** : très grand volume de données ou nombreuses variables (calcul lent, distances peu informatives), ou variables d'échelles et de pertinences très différentes.
- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC ; $k$ choisi par validation croisée.

### Analyse de l'expérience Python
- **Entrées** : `Xc_ent`, `yc_ent` pour l'entraînement ; `Xc_test`, `yc_test` pour l'évaluation. **Hyperparamètres** : `n_neighbors=5` ; `StandardScaler` dans un pipeline pour mettre les variables à la même échelle ; autres options par défaut.
- **Sortie** : F1 = 0,8889, moyenne harmonique de la précision et du rappel pour la classe positive.
- **Ce que cela permet** : si le jeu de test est le même que pour la régression logistique, constater que ce F1 est supérieur (0,8113) : le KNN classe mieux ici, ce qui peut suggérer une frontière non linéaire.
- **Ce que cela ne permet pas** : affirmer que le KNN est réellement meilleur, car l'écart provient d'une seule découpe train/test (surtout si le jeu de test est petit) ; connaître la précision, le rappel et la matrice de confusion séparément ; savoir si $k=5$ est optimal (valeur fixée, non validée).

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
- **Nom** : Classifieur bayésien naïf (ici gaussien) – *Naive Bayes (Gaussian NB)*. **Famille** : classification supervisée (modèle probabiliste) ; objectif : choisir la classe la plus probable sachant les variables observées.
- **Intuition** : on applique le théorème de Bayes en supposant, de façon « naïve », que les variables sont indépendantes entre elles à l'intérieur de chaque classe.
- **Étapes** : estimation, pour chaque classe, de sa fréquence et de la moyenne et variance de chaque variable → calcul de la probabilité de chaque classe pour une nouvelle observation → choix de la classe de probabilité maximale.
- **Critère** : $\hat{y}=\underset{c}{\arg\max}\ P(c)\prod_{j=1}^{p}\mathcal{N}\left(x_j\mid\mu_{cj},\sigma_{cj}^2\right)$
- **Symboles** : $c$ = classe ; $P(c)$ = probabilité a priori de la classe ; $x_j$ = variable $j$ ; $p$ = nombre de variables ; $\mathcal{N}$ = densité gaussienne ; $\mu_{cj},\sigma_{cj}^2$ = moyenne et variance de la variable $j$ dans la classe $c$.
- **Paramètres** (appris) : $P(c)$, $\mu_{cj}$ et $\sigma_{cj}^2$. **Hyperparamètres** : pratiquement aucun (ici valeurs par défaut, dont le lissage de variance `var_smoothing`).
- **Avantage 1** : très rapide, fonctionne avec peu de données et beaucoup de variables.
- **Avantage 2** : fournit des probabilités et reste simple à mettre en œuvre.
- **Limite 1** : l'hypothèse d'indépendance est rarement vraie (variables corrélées) et peut dégrader les prédictions.
- **Limite 2** : la version gaussienne suppose des variables à distribution approximativement normale par classe ; les probabilités produites sont souvent mal calibrées.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

rng = np.random.default_rng(0)
x0 = rng.normal(-1, 1.0, 100)   # classe 0 (données synthétiques)
x1 = rng.normal(1.5, 1.2, 100)  # classe 1
xs = np.linspace(-5, 6, 400)

d0 = 0.5 * norm.pdf(xs, x0.mean(), x0.std())
d1 = 0.5 * norm.pdf(xs, x1.mean(), x1.std())

plt.figure(figsize=(6, 4))
plt.hist(x0, bins=20, density=True, alpha=0.3, color="tab:blue")
plt.hist(x1, bins=20, density=True, alpha=0.3, color="tab:orange")
plt.plot(xs, d0, color="tab:blue", label="P(classe 0) × N(x | μ₀, σ₀²)")
plt.plot(xs, d1, color="tab:orange", label="P(classe 1) × N(x | μ₁, σ₁²)")
frontiere = xs[np.argmin(np.abs(d0 - d1) + (xs < x0.mean()) * 10 + (xs > x1.mean()) * 10)]
plt.axvline(frontiere, color="red", ls="--", label="Frontière de décision")
plt.xlabel("Variable explicative x"); plt.ylabel("Densité")
plt.title("Naive Bayes gaussien : on choisit la classe la plus probable (données synthétiques)")
plt.legend(fontsize=8); plt.show()
```

**Lecture du schéma** : les histogrammes montrent la répartition réelle de x dans chaque classe. Les courbes sont les lois gaussiennes estimées par le modèle (moyenne et variance de chaque classe), pondérées par la fréquence de la classe. Pour une valeur de x, le modèle choisit la classe dont la courbe est la plus haute ; la ligne rouge pointillée marque le point où les deux courbes se croisent.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/commerce)** : filtrer des messages ou avis clients (spam ou non, avis positif ou négatif) à partir de la fréquence des mots (variante multinomiale).
- **Cas d'usage 2 (finance/gestion)** : pré-classer rapidement des dossiers (risque faible ou élevé) à partir de quelques indicateurs numériques, quand les données d'entraînement sont peu nombreuses.
- **Données et prérequis** : variables numériques (version gaussienne) ou comptages (version multinomiale) ; pas de standardisation nécessaire ; effectif même réduit.
- **Cas peu approprié** : variables fortement corrélées entre elles, ou lorsque des probabilités bien calibrées sont indispensables (par exemple pour fixer un prix ou un seuil de risque).
- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC ; log-loss si l'on s'intéresse aux probabilités.

### Analyse de l'expérience Python
- **Entrées** : `Xc_ent`, `yc_ent` pour l'entraînement ; `Xc_test`, `yc_test` pour l'évaluation. **Hyperparamètres** : aucun renseigné, `GaussianNB()` avec ses valeurs par défaut ; pas de standardisation.
- **Sortie** : F1 = 0,7551, moyenne harmonique de la précision et du rappel pour la classe positive.
- **Ce que cela permet** : si le jeu de test est le même que pour les modèles 8 et 9, constater que ce F1 est inférieur à celui de la régression logistique (0,8113) et du KNN (0,8889) ; l'hypothèse d'indépendance ou de normalité est peut-être peu adaptée à ces données.
- **Ce que cela ne permet pas** : confirmer cette explication sans analyse supplémentaire (corrélations, distributions) ; connaître la précision, le rappel et la matrice de confusion séparément ; conclure de façon définitive à partir d'une seule découpe train/test.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
- **Nom** : Arbre de décision (CART) pour la classification – *Decision Tree Classifier*. **Famille** : classification supervisée ; objectif : prédire une classe par une suite de règles « si… alors… ».
- **Intuition** : on découpe l'espace des variables en zones de plus en plus homogènes, chaque zone (feuille) prédisant la classe majoritaire de ses observations.
- **Étapes** : données (X, y) → recherche récursive de la variable et du seuil qui séparent le mieux les classes → arrêt à la profondeur maximale → prédiction = classe majoritaire de la feuille atteinte.
- **Critère** : à chaque nœud, on minimise l'impureté de Gini pondérée des deux groupes : $G=1-\sum_{c=1}^{C}p_c^2$.
- **Symboles** : $C$ = nombre de classes ; $p_c$ = proportion d'observations de la classe $c$ dans le nœud ($G=0$ si le nœud est pur).
- **Paramètres** (appris) : variables et seuils de coupure, classes des feuilles. **Hyperparamètres** : `max_depth` (ici 4), `random_state` (ici `GRAINE`) ; critère (Gini) et autres options par défaut.
- **Avantage 1** : très lisible (règles visualisables), sans standardisation nécessaire.
- **Avantage 2** : capte des relations non linéaires et des interactions automatiquement.
- **Limite 1** : instable (un petit changement des données peut modifier l'arbre) et sujet au surapprentissage si trop profond.
- **Limite 2** : frontières de décision en « escalier » (parallèles aux axes), peu adaptées aux frontières obliques.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree

rng = np.random.default_rng(0)
A = rng.normal([-1, -1], 1.0, (60, 2))     # classe 0 (données synthétiques)
B = rng.normal([1.5, 1.5], 1.0, (60, 2))   # classe 1
X = np.vstack([A, B]); y = np.array([0]*60 + [1]*60)
arbre = DecisionTreeClassifier(max_depth=2, random_state=0).fit(X, y)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
# Gauche : zones de décision
xx, yy = np.meshgrid(np.linspace(-4, 5, 300), np.linspace(-4, 5, 300))
zz = arbre.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
axes[0].contourf(xx, yy, zz, alpha=0.25, cmap="coolwarm")
axes[0].scatter(*A.T, color="tab:blue", s=15, label="Classe 0")
axes[0].scatter(*B.T, color="tab:red", s=15, label="Classe 1")
axes[0].set_xlabel("Variable 1"); axes[0].set_ylabel("Variable 2")
axes[0].set_title("Zones de décision (rectangles)"); axes[0].legend()
# Droite : structure de l'arbre
plot_tree(arbre, feature_names=["Variable 1", "Variable 2"],
          class_names=["Classe 0", "Classe 1"], filled=True, ax=axes[1])
axes[1].set_title("Règles de l'arbre (profondeur 2)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : à gauche, le plan est découpé en rectangles colorés : chaque couleur correspond à la classe prédite dans cette zone. À droite, le même modèle sous forme de règles : on part du nœud du haut, on répond à chaque test (« Variable ≤ seuil ? ») en descendant à gauche si oui, à droite si non, jusqu'à une feuille qui donne la classe. Les couleurs indiquent la classe majoritaire, et l'intensité reflète la pureté du nœud.

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : décider d'accorder ou non un crédit selon des règles claires (revenu, endettement, ancienneté), faciles à expliquer au client et au régulateur.
- **Cas d'usage 2 (marketing/logistique)** : classer les clients selon leur propension à acheter, ou les commandes selon qu'elles risquent d'être livrées en retard ou non.
- **Données et prérequis** : variables numériques ou catégorielles encodées ; pas de standardisation ; effectif suffisant pour éviter des feuilles quasi vides.
- **Cas peu approprié** : frontière entre classes oblique ou très lisse, ou petit jeu de données avec beaucoup de bruit (l'arbre mémorise le bruit).
- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC ; profondeur choisie par validation croisée.

### Analyse de l'expérience Python
- **Entrées** : `Xc_ent`, `yc_ent` pour l'entraînement ; `Xc_test`, `yc_test` pour l'évaluation. **Hyperparamètres** : `max_depth=4`, `random_state=GRAINE` ; critère Gini et autres options par défaut.
- **Sortie** : F1 = 0,8911, moyenne harmonique de la précision et du rappel pour la classe positive.
- **Ce que cela permet** : si le jeu de test est le même que pour les modèles 8 à 10, constater que ce F1 est supérieur à celui de la régression logistique (0,8113) et de Naive Bayes (0,7551), et à peu près équivalent à celui du KNN (0,8889).
- **Ce que cela ne permet pas** : affirmer que l'arbre est meilleur que le KNN (écart d'environ 0,002, négligeable sur une seule découpe train/test) ; connaître la précision, le rappel et la matrice de confusion séparément ; savoir si une profondeur de 4 est optimale (valeur fixée, non validée) ni si l'arbre surapprend (score d'entraînement non affiché).

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import RandomForestClassifier
modele=RandomForestClassifier(n_estimators=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
- **Nom** : Forêt aléatoire pour la classification – *Random Forest Classifier*. **Famille** : classification supervisée (méthode d'ensemble) ; objectif : prédire une classe par vote de nombreux arbres.
- **Intuition** : un seul arbre est instable ; en faisant voter plusieurs arbres entraînés sur des échantillons et des variables différents, les erreurs individuelles se compensent.
- **Étapes** : tirage de $B$ échantillons bootstrap (avec remise) → entraînement d'un arbre par échantillon, en ne considérant à chaque coupure qu'un sous-ensemble aléatoire de variables → prédiction = classe majoritaire parmi les $B$ arbres.
- **Critère** : $\hat{y}(x)=\underset{c}{\arg\max}\sum_{b=1}^{B}\mathbb{1}\left[T_b(x)=c\right]$
- **Symboles** : $B$ = nombre d'arbres (`n_estimators`) ; $T_b(x)$ = classe prédite par le $b$-ième arbre ; $c$ = classe candidate ; $\mathbb{1}[\cdot]$ = 1 si la condition est vraie, 0 sinon.
- **Paramètres** (appris) : seuils et feuilles de chaque arbre. **Hyperparamètres** : `n_estimators` (ici 80), `random_state` (ici `GRAINE`) ; `max_depth`, `max_features`, etc. à leurs valeurs par défaut.
- **Avantage 1** : bien plus stable et moins sujette au surapprentissage qu'un arbre isolé.
- **Avantage 2** : peu de réglages nécessaires, gère les non-linéarités et fournit l'importance des variables.
- **Limite 1** : moins interprétable (80 arbres au lieu d'un) et plus coûteuse en calcul et en mémoire.
- **Limite 2** : probabilités parfois mal calibrées et performances limitées sur des classes très déséquilibrées sans ajustement.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

rng = np.random.default_rng(0)
A = rng.normal([-1, -1], 1.1, (60, 2))     # classe 0 (données synthétiques)
B = rng.normal([1.2, 1.2], 1.1, (60, 2))   # classe 1
X = np.vstack([A, B]); y = np.array([0]*60 + [1]*60)
xx, yy = np.meshgrid(np.linspace(-4, 4.5, 250), np.linspace(-4, 4.5, 250))
grille = np.c_[xx.ravel(), yy.ravel()]

fig, axes = plt.subplots(1, 4, figsize=(15, 3.5), sharey=True)
votes = np.zeros(len(grille))
for b in range(25):
    idx = rng.integers(0, len(X), len(X))          # échantillon bootstrap
    arbre = DecisionTreeClassifier(max_depth=4, random_state=b).fit(X[idx], y[idx])
    votes += arbre.predict(grille)
    if b < 3:
        axes[b].contourf(xx, yy, arbre.predict(grille).reshape(xx.shape), alpha=0.25, cmap="coolwarm")
        axes[b].scatter(*A.T, color="tab:blue", s=8); axes[b].scatter(*B.T, color="tab:red", s=8)
        axes[b].set_title(f"Arbre {b+1}"); axes[b].set_xlabel("Variable 1")
axes[3].contourf(xx, yy, (votes / 25 > 0.5).reshape(xx.shape), alpha=0.25, cmap="coolwarm")
axes[3].scatter(*A.T, color="tab:blue", s=8); axes[3].scatter(*B.T, color="tab:red", s=8)
axes[3].set_title("Vote majoritaire (25 arbres)"); axes[3].set_xlabel("Variable 1")
axes[0].set_ylabel("Variable 2")
fig.suptitle("Forêt aléatoire : agrégation par vote de plusieurs arbres (données synthétiques)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : les trois premiers graphiques montrent des arbres entraînés sur des échantillons bootstrap différents : leurs zones de décision (bleu = classe 0, rouge = classe 1) sont irrégulières et diffèrent d'un arbre à l'autre. Le dernier graphique montre la décision de la forêt, obtenue par vote majoritaire de 25 arbres : la frontière est plus régulière et plus stable que celle de chaque arbre pris isolément.

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : détecter des fraudes ou prédire un défaut de paiement à partir de nombreuses variables hétérogènes (montants, fréquence, ancienneté, localisation).
- **Cas d'usage 2 (marketing/commerce)** : prédire le désabonnement (*churn*) ou la réponse à une campagne à partir de l'historique client.
- **Données et prérequis** : tableau de variables numériques ou catégorielles encodées, cible catégorielle, effectif moyen à grand ; pas de standardisation nécessaire.
- **Cas peu approprié** : contexte exigeant des règles simples et transparentes (justification à un régulateur), ou données très haute dimension et creuses (texte brut), où des modèles linéaires sont souvent plus efficaces.
- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC ; erreur *out-of-bag* ou validation croisée pour le réglage.

### Analyse de l'expérience Python
- **Entrées** : `Xc_ent`, `yc_ent` pour l'entraînement ; `Xc_test`, `yc_test` pour l'évaluation. **Hyperparamètres** : `n_estimators=80`, `random_state=GRAINE` ; les autres gardent leurs valeurs par défaut.
- **Sortie** : F1 = 0,9184, moyenne harmonique de la précision et du rappel pour la classe positive.
- **Ce que cela permet** : si le jeu de test est le même que pour les modèles 8 à 11, constater que c'est le meilleur F1 obtenu jusqu'ici, supérieur à celui de l'arbre seul (0,8911), du KNN (0,8889), de la régression logistique (0,8113) et de Naive Bayes (0,7551), ce qui est cohérent avec l'effet stabilisateur du vote.
- **Ce que cela ne permet pas** : conclure à une supériorité nette, car l'écart avec l'arbre (environ 0,03) provient d'une seule découpe train/test, probablement de petite taille ; connaître la précision, le rappel et la matrice de confusion séparément ; savoir si un autre nombre d'arbres ou une autre profondeur donnerait mieux (non testé).

### Présentation
- **Nom** : Machine à vecteurs de support (ici à noyau gaussien) – *Support Vector Machine (SVM, SVC)*. **Famille** : classification supervisée ; objectif : trouver la frontière qui sépare les classes avec la marge la plus grande possible.
- **Intuition** : parmi toutes les frontières possibles, on retient celle la plus éloignée des points les plus proches de chaque classe (les vecteurs de support) ; le noyau permet de courber cette frontière.
- **Étapes** : standardisation → apprentissage des vecteurs de support et de leurs poids (problème d'optimisation convexe) → prédiction de la classe selon le côté de la frontière où tombe la nouvelle observation.
- **Critère** : $\min_{w,b,\xi}\ \frac{1}{2}\|w\|^2+C\sum_{i=1}^{n}\xi_i$ sous $y_i\,(w^\top\phi(x_i)+b)\ge 1-\xi_i$, avec le noyau $K(x,x')=\exp\left(-\gamma\|x-x'\|^2\right)$.
- **Symboles** : $w$ = vecteur normal à la frontière ; $b$ = biais ; $\xi_i\ge 0$ = erreur tolérée pour l'observation $i$ ; $y_i\in\{-1,+1\}$ = classe ; $\phi$ = transformation implicite du noyau ; $\gamma$ = largeur du noyau.
- **Paramètres** (appris) : vecteurs de support et leurs coefficients, $b$. **Hyperparamètres** : `kernel` (ici `'rbf'`), `C` (ici 1,0, compromis marge/erreurs), `random_state` (ici `GRAINE`) ; $\gamma$ à sa valeur par défaut.
- **Avantage 1** : efficace en dimension élevée et capable de frontières non linéaires grâce au noyau.
- **Avantage 2** : repose sur peu de points (les vecteurs de support), robuste quand la marge est nette.
- **Limite 1** : coût de calcul élevé sur de très grands jeux de données et sensibilité à l'échelle des variables (d'où la standardisation).
- **Limite 2** : ne fournit pas directement de probabilités, et le résultat est peu interprétable ; sensible au choix de `C` et $\gamma$.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC

rng = np.random.default_rng(0)
A = rng.normal([-1.5, -1.5], 0.9, (40, 2))   # classe 0 (données synthétiques)
B = rng.normal([1.5, 1.5], 0.9, (40, 2))     # classe 1
X = np.vstack([A, B]); y = np.array([0]*40 + [1]*40)
svm = SVC(kernel="linear", C=1.0).fit(X, y)

xx, yy = np.meshgrid(np.linspace(-4.5, 4.5, 300), np.linspace(-4.5, 4.5, 300))
z = svm.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6, 5))
plt.scatter(*A.T, color="tab:blue", label="Classe 0")
plt.scatter(*B.T, color="tab:orange", label="Classe 1")
plt.contour(xx, yy, z, levels=[0], colors="red", linewidths=2)
plt.contour(xx, yy, z, levels=[-1, 1], colors="gray", linestyles="--")
plt.scatter(*svm.support_vectors_.T, s=140, facecolors="none", edgecolors="black", label="Vecteurs de support")
plt.plot([], [], color="red", label="Frontière de décision")
plt.plot([], [], color="gray", ls="--", label="Bords de la marge")
plt.xlabel("Variable 1"); plt.ylabel("Variable 2")
plt.title("SVM : frontière à marge maximale (données synthétiques, noyau linéaire)")
plt.legend(fontsize=8); plt.show()
```

**Lecture du schéma** : les points bleus et orange sont les deux classes. La ligne rouge est la frontière de décision ; les deux lignes grises en pointillés délimitent la marge, c'est-à-dire la bande vide la plus large possible entre les classes. Les points entourés en noir sont les vecteurs de support : ils sont les seuls à déterminer la position de la frontière. Pour simplifier la lecture, le schéma utilise un noyau linéaire ; le noyau `'rbf'` du code produit la même logique avec une frontière courbe.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion)** : classer des entreprises ou des dossiers (saines ou en difficulté) à partir d'un grand nombre de ratios financiers.

- **Cas peu approprié** : très grand jeu de données (temps d'entraînement élevé) ou besoin de probabilités bien calibrées et de règles explicables.

### Présentation
- **Nom** : Gradient Boosting pour la classification (ici version par histogrammes) – *Histogram-based Gradient Boosting Classifier*. **Famille** : classification supervisée (méthode d'ensemble) ; objectif : prédire une classe en corrigeant progressivement les erreurs des arbres précédents.
- **Intuition** : on construit les arbres un par un ; chaque nouvel arbre apprend à corriger les erreurs commises par l'ensemble des arbres précédents, en suivant le gradient de la fonction de perte.
- **Étapes** : score initial (log-odds de la classe) → calcul du gradient de la perte logistique → ajustement d'un petit arbre sur ce gradient → ajout de sa contribution, pondérée, au score → répétition `max_iter` fois → conversion du score en probabilité puis en classe.
- **Critère** : $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$, avec $P(y=1\mid x)=\dfrac{1}{1+e^{-F_M(x)}}$ ; $h_m$ ajuste le gradient négatif de la perte logistique.
- **Symboles** : $F_m$ = score après $m$ arbres ; $h_m$ = $m$-ième arbre ; $\eta$ = taux d'apprentissage (`learning_rate`) ; $M$ = nombre total d'arbres.
- **Paramètres** (appris) : structure et feuilles de chaque arbre. **Hyperparamètres** : `max_iter` (ici 80, nombre d'arbres), `random_state` (ici `GRAINE`) ; `learning_rate`, `max_depth`, etc. à leurs valeurs par défaut.
- **Avantage 1** : souvent parmi les modèles les plus précis sur données tabulaires, avec non-linéarités et interactions captées.
- **Avantage 2** : la version par histogrammes est rapide et accepte les valeurs manquantes.
- **Limite 1** : sensible au réglage (nombre d'arbres, taux d'apprentissage) et risque de surapprentissage.
- **Limite 2** : peu interprétable, et plus difficile à calibrer et à expliquer qu'un modèle linéaire.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

rng = np.random.default_rng(0)
A = rng.normal([-1, -1], 1.0, (60, 2))     # classe 0 (données synthétiques)
B = rng.normal([1.2, 1.2], 1.0, (60, 2))   # classe 1
X = np.vstack([A, B]); y = np.array([0]*60 + [1]*60)
xx, yy = np.meshgrid(np.linspace(-4, 4.5, 250), np.linspace(-4, 4.5, 250))
grille = np.c_[xx.ravel(), yy.ravel()]

eta = 0.3
F = np.zeros(len(X)); Fg = np.zeros(len(grille))
etapes = [1, 3, 10, 40]
fig, axes = plt.subplots(1, 4, figsize=(15, 3.5), sharey=True)
k = 0
for m in range(1, max(etapes) + 1):
    p = 1 / (1 + np.exp(-F))
    arbre = DecisionTreeRegressor(max_depth=2).fit(X, y - p)   # arbre sur le gradient (y - p)
    F += eta * arbre.predict(X)
    Fg += eta * arbre.predict(grille)
    if m in etapes:
        ax = axes[k]; k += 1
        proba = 1 / (1 + np.exp(-Fg))
        ax.contourf(xx, yy, proba.reshape(xx.shape), levels=np.linspace(0, 1, 11), alpha=0.35, cmap="coolwarm")
        ax.scatter(*A.T, color="tab:blue", s=8); ax.scatter(*B.T, color="tab:red", s=8)
        ax.set_title(f"{m} arbre(s)"); ax.set_xlabel("Variable 1")
axes[0].set_ylabel("Variable 2")
fig.suptitle("Gradient Boosting : la probabilité prédite s'affine à chaque arbre ajouté (données synthétiques)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : de gauche à droite, on ajoute des arbres. Les couleurs de fond représentent la probabilité prédite d'appartenir à la classe 1 (bleu = faible, rouge = forte) ; les points sont les observations réelles. Avec un seul arbre, la carte est grossière et peu contrastée ; à mesure que les arbres s'ajoutent, chacun corrigeant les erreurs restantes, la séparation entre les deux classes devient plus nette et plus précise.

### Contextes d'utilisation

- **Cas d'usage 2 (marketing/logistique)** : prédire le désabonnement d'un client ou le risque de retard de livraison à partir de variables de prix, de lieu, de période et d'historique.

- **Cas peu approprié** : très petit jeu de données (surapprentissage) ou contexte exigeant des règles simples et transparentes.
- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC, log-loss ; réglage par validation croisée ou arrêt précoce sur un jeu de validation.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
- **Nom** : AdaBoost (Boosting adaptatif) – *Adaptive Boosting (AdaBoost)*. **Famille** : classification supervisée (méthode d'ensemble) ; objectif : combiner de nombreux classifieurs faibles en un classifieur fort.
- **Intuition** : on entraîne des modèles très simples (par défaut des « souches », arbres à un seul test) l'un après l'autre ; après chaque tour, on augmente le poids des observations mal classées pour que le modèle suivant se concentre dessus.
- **Étapes** : poids égaux pour toutes les observations → entraînement d'un classifieur faible pondéré → calcul de son erreur et de son poids dans le vote → mise à jour des poids des observations → répétition `n_estimators` fois → vote pondéré final.
- **Critère** : $H(x)=\operatorname{signe}\left(\sum_{m=1}^{M}\alpha_m\,h_m(x)\right)$ avec $\alpha_m=\frac{1}{2}\ln\frac{1-\varepsilon_m}{\varepsilon_m}$.
- **Symboles** : $M$ = nombre de classifieurs faibles ; $h_m(x)\in\{-1,+1\}$ = prédiction du $m$-ième classifieur ; $\varepsilon_m$ = son erreur pondérée ; $\alpha_m$ = son poids dans le vote (plus il est précis, plus il compte).
- **Paramètres** (appris) : classifieurs faibles et leurs poids $\alpha_m$. **Hyperparamètres** : `n_estimators` (ici 70), `random_state` (ici `GRAINE`) ; `learning_rate` et classifieur de base à leurs valeurs par défaut.
- **Avantage 1** : simple et efficace, améliore nettement des modèles faibles, peu de réglages.
- **Avantage 2** : peu sujet au surapprentissage avec des modèles de base simples.
- **Limite 1** : sensible aux valeurs aberrantes et au bruit dans les étiquettes (leurs poids ne cessent d'augmenter).
- **Limite 2** : entraînement séquentiel (non parallélisable) et résultats souvent dépassés par le gradient boosting.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

rng = np.random.default_rng(0)
A = rng.normal([-1, -1], 1.0, (40, 2))     # classe 0 (données synthétiques)
B = rng.normal([1, 1], 1.0, (40, 2))       # classe 1
X = np.vstack([A, B]); y = np.array([0]*40 + [1]*40)
ys = 2*y - 1                                # étiquettes -1 / +1
xx, yy = np.meshgrid(np.linspace(-4, 4, 250), np.linspace(-4, 4, 250))
grille = np.c_[xx.ravel(), yy.ravel()]

w = np.full(len(X), 1/len(X))
score = np.zeros(len(grille))
fig, axes = plt.subplots(1, 4, figsize=(15, 3.5), sharey=True)
for m in range(3):
    souche = DecisionTreeClassifier(max_depth=1).fit(X, y, sample_weight=w)
    pred = 2*souche.predict(X) - 1
    err = w[pred != ys].sum()
    alpha = 0.5*np.log((1 - err)/err)
    score += alpha * (2*souche.predict(grille) - 1)
    axes[m].contourf(xx, yy, souche.predict(grille).reshape(xx.shape), alpha=0.2, cmap="coolwarm")
    axes[m].scatter(*X.T, c=y, cmap="coolwarm", s=10 + 400*w, edgecolors="k", linewidths=0.3)
    axes[m].set_title(f"Tour {m+1} : souche"); axes[m].set_xlabel("Variable 1")
    w = w * np.exp(-alpha * ys * pred); w /= w.sum()     # mise à jour des poids
axes[3].contourf(xx, yy, (score > 0).reshape(xx.shape), alpha=0.2, cmap="coolwarm")
axes[3].scatter(*X.T, c=y, cmap="coolwarm", s=10, edgecolors="k", linewidths=0.3)
axes[3].set_title("Vote pondéré des 3 souches"); axes[3].set_xlabel("Variable 1")
axes[0].set_ylabel("Variable 2")
fig.suptitle("AdaBoost : les points mal classés grossissent, le tour suivant s'y concentre (données synthétiques)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : les trois premiers graphiques montrent les tours successifs. Chaque fond coloré est la décision d'une souche (un seul seuil, donc une frontière droite) ; la taille d'un point est proportionnelle à son poids. Après chaque tour, les points mal classés grossissent, ce qui oblige la souche suivante à s'en occuper. Le dernier graphique montre le vote pondéré des trois souches, dont la frontière est plus fine que celle de chacune.

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : classer des dossiers de crédit (accepté ou refusé) à partir de quelques indicateurs, en combinant de nombreuses règles simples.
- **Cas d'usage 2 (marketing/commerce)** : prédire si un client répondra à une campagne à partir de son profil et de son historique.
- **Données et prérequis** : données tabulaires numériques ou catégorielles encodées, cible binaire ou multiclasse, étiquettes fiables ; pas de standardisation nécessaire.
- **Cas peu approprié** : données bruitées avec beaucoup d'erreurs d'étiquetage ou de valeurs aberrantes, ou très grand volume exigeant un entraînement parallèle.
- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC ; `n_estimators` et `learning_rate` choisis par validation croisée.

### Analyse de l'expérience Python
- **Entrées** : `Xc_ent`, `yc_ent` pour l'entraînement ; `Xc_test`, `yc_test` pour l'évaluation. **Hyperparamètres** : `n_estimators=70`, `random_state=GRAINE` ; les autres gardent leurs valeurs par défaut.
- **Sortie** : F1 = 0,8269, moyenne harmonique de la précision et du rappel pour la classe positive.
- **Ce que cela permet** : si le jeu de test est le même que pour les modèles 8 à 14, constater que ce F1 est supérieur à ceux de la régression logistique (0,8113) et de Naive Bayes (0,7551), mais inférieur à ceux du KNN (0,8889), de l'arbre (0,8911), du SVM (0,8980), de la forêt (0,9184) et du gradient boosting (0,96).

```python
from xgboost import XGBClassifier
modele=XGBClassifier(n_estimators=80,max_depth=3,random_state=GRAINE,eval_metric='logloss').fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
- **Nom** : XGBoost (Extreme Gradient Boosting) – *eXtreme Gradient Boosting*. **Famille** : classification supervisée (méthode d'ensemble, bibliothèque externe `xgboost`) ; objectif : prédire une classe par un ensemble d'arbres construits séquentiellement avec une perte régularisée.
- **Intuition** : comme le gradient boosting, chaque arbre corrige les erreurs des précédents, mais XGBoost utilise en plus la dérivée seconde de la perte et une pénalité sur la complexité des arbres.
- **Étapes** : score initial → calcul du gradient et de la hessienne de la perte logistique → construction d'un arbre de profondeur limitée qui maximise le gain régularisé → ajout de sa contribution, pondérée, au score → répétition `n_estimators` fois → conversion en probabilité puis en classe.
- **Critère** : $\mathcal{L}=\sum_{i=1}^{n}\ell\left(y_i,\hat{y}_i\right)+\sum_{m=1}^{M}\Omega(f_m)$ avec $\Omega(f)=\gamma\,T+\frac{1}{2}\lambda\sum_{k=1}^{T}w_k^2$.
- **Symboles** : $\ell$ = perte (ici logistique, `logloss`) ; $f_m$ = $m$-ième arbre ; $T$ = nombre de feuilles ; $w_k$ = valeur de la feuille $k$ ; $\gamma,\lambda$ = pénalités sur le nombre de feuilles et sur leurs valeurs.
- **Paramètres** (appris) : structure des arbres et valeurs des feuilles. **Hyperparamètres** : `n_estimators` (ici 80), `max_depth` (ici 3), `random_state` (ici `GRAINE`), `eval_metric='logloss'` ; `learning_rate`, $\gamma$, $\lambda$ à leurs valeurs par défaut.
- **Avantage 1** : très performant sur données tabulaires, avec régularisation intégrée contre le surapprentissage.
- **Avantage 2** : rapide (parallélisation) et gère les valeurs manquantes.
- **Limite 1** : nombreux hyperparamètres à régler et sensible à ce réglage.
- **Limite 2** : peu interprétable et dépendance à une bibliothèque externe à installer.

**Lecture du schéma** : on lit de gauche à droite. Les scores de tous les arbres s'additionnent pour donner la prédiction finale (la classe). La phrase en haut rappelle ce qui distingue XGBoost : à chaque étape, il utilise la pente et la courbure de la perte, et pénalise la complexité des arbres.

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : détection de fraude ou scoring de crédit sur de grands volumes de transactions et de nombreuses variables.

- **Métriques adaptées** : F1, précision, rappel, matrice de confusion, AUC-ROC, log-loss ; hyperparamètres réglés par validation croisée avec arrêt précoce.

- **Entrées** : `Xc_ent`, `yc_ent` pour l'entraînement ; `Xc_test`, `yc_test` pour l'évaluation. **Hyperparamètres** : `n_estimators=80`, `max_depth=3`, `random_state=GRAINE`, `eval_metric='logloss'` ; les autres gardent leurs valeurs par défaut.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

### Présentation
- **Nom** : K-moyennes – *K-means*. **Famille** : clustering (apprentissage non supervisé) ; objectif : regrouper les observations en $k$ groupes compacts et homogènes.
- **Intuition** : chaque groupe est représenté par son centre (centroïde) ; chaque point est affecté au centre le plus proche, puis les centres sont recalculés.
- **Étapes** : choix de $k$ centres initiaux → affectation de chaque point au centroïde le plus proche → recalcul de chaque centroïde (moyenne des points du groupe) → répétition jusqu'à stabilisation → sortie : une étiquette de groupe par observation.
- **Critère** : $\min_{C,\mu}\ \sum_{j=1}^{k}\sum_{x_i\in C_j}\|x_i-\mu_j\|^2$
- **Symboles** : $k$ = nombre de groupes ; $C_j$ = groupe $j$ ; $\mu_j$ = centroïde du groupe $j$ ; $x_i$ = observation $i$ ; $\|\cdot\|$ = distance euclidienne.
- **Paramètres** (appris) : positions des centroïdes $\mu_j$ et étiquettes. **Hyperparamètres** : `n_clusters` (ici 3), `n_init` (ici 10, nombre d'initialisations testées), `random_state` (ici `GRAINE`).
- **Avantage 1** : simple, rapide, adapté à de grands jeux de données.
- **Avantage 2** : résultat facile à interpréter (un centre représentatif par groupe).
- **Limite 1** : $k$ doit être fixé à l'avance et le résultat dépend de l'initialisation.
- **Limite 2** : suppose des groupes compacts, de forme sphérique et de tailles proches ; sensible aux valeurs aberrantes et à l'échelle des variables.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
X = np.vstack([rng.normal([0, 0], 0.7, (40, 2)),
               rng.normal([4, 4], 0.7, (40, 2)),
               rng.normal([0, 4.5], 0.7, (40, 2))])   # données synthétiques
mu = X[rng.choice(len(X), 3, replace=False)]            # centroïdes initiaux

fig, axes = plt.subplots(1, 4, figsize=(15, 3.5), sharex=True, sharey=True)
couleurs = ["tab:blue", "tab:orange", "tab:green"]
for it in range(4):
    d = np.linalg.norm(X[:, None, :] - mu[None, :, :], axis=2)
    lab = d.argmin(axis=1)                               # affectation
    for j in range(3):
        axes[it].scatter(*X[lab == j].T, color=couleurs[j], s=12)
    axes[it].scatter(*mu.T, color="red", marker="X", s=140, edgecolors="black")
    axes[it].set_title(f"Itération {it+1}"); axes[it].set_xlabel("Variable 1")
    mu = np.array([X[lab == j].mean(axis=0) if (lab == j).any() else mu[j] for j in range(3)])
axes[0].set_ylabel("Variable 2")
fig.suptitle("K-means : affectation au centroïde le plus proche, puis mise à jour des centroïdes (données synthétiques)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : chaque graphique est une itération. Les croix rouges sont les centroïdes ; la couleur d'un point indique le groupe auquel il est affecté (centroïde le plus proche). D'une itération à l'autre, les croix se déplacent vers le centre de leur groupe et les couleurs se stabilisent : l'algorithme a convergé quand plus rien ne change.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing)** : segmenter la clientèle (fréquence d'achat, panier moyen, ancienneté) pour adapter les offres.
- **Cas d'usage 2 (logistique/commerce)** : regrouper des points de livraison ou des magasins par proximité géographique ou par profil de ventes.
- **Données et prérequis** : variables numériques standardisées, nombre de groupes $k$ choisi à l'avance, pas de valeurs manquantes.
- **Cas peu approprié** : groupes de forme allongée ou imbriquée (par exemple deux anneaux concentriques), de tailles très différentes, ou données avec beaucoup de points aberrants.
- **Métriques adaptées** : coefficient de silhouette, inertie (méthode du coude), indice de Davies-Bouldin ; validation par l'utilité métier des groupes, car il n'y a pas de vérité terrain.

### Analyse de l'expérience Python
- **Entrées** : `X_clu` (données sans étiquettes, pas de découpe train/test). **Hyperparamètres** : `n_clusters=3`, `n_init=10`, `random_state=GRAINE`. La sortie comprend le score de silhouette et un nuage de points coloré selon `modele.labels_`.
- **Sortie** : silhouette = 0,8686. Ce coefficient va de −1 à 1 : plus il est proche de 1, mieux les points sont proches de leur groupe et éloignés des autres groupes.
- **Ce que cela permet** : constater que, pour $k=3$, les groupes obtenus sont bien séparés et compacts selon ce critère (valeur élevée).
- **Ce que cela ne permet pas** : prouver que $k=3$ est le bon nombre de groupes (aucun autre $k$ n'a été comparé) ; garantir que les groupes ont un sens métier ; comparer à une vérité terrain (il n'y en a pas) ; exclure un autre minimum local lié à l'initialisation. La silhouette favorise par ailleurs les groupes convexes.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Présentation
- **Nom** : Classification ascendante hiérarchique (CAH) – *Agglomerative (Hierarchical) Clustering*. **Famille** : clustering (apprentissage non supervisé) ; objectif : construire une hiérarchie de regroupements, puis la couper pour obtenir un nombre voulu de groupes.
- **Intuition** : on part de chaque point isolé et on fusionne, pas à pas, les deux groupes les plus proches, jusqu'à n'en avoir plus qu'un seul ; on « coupe » ensuite l'arbre de fusions au niveau souhaité.
- **Étapes** : chaque observation = un groupe → calcul des distances entre groupes → fusion des deux plus proches → répétition jusqu'à obtenir `n_clusters` groupes → sortie : une étiquette par observation.
- **Critère** : on fusionne les groupes $A$ et $B$ qui minimisent l'augmentation d'inertie intra-groupe (critère de Ward, par défaut dans scikit-learn) : $\Delta(A,B)=\frac{n_An_B}{n_A+n_B}\|\mu_A-\mu_B\|^2$.
- **Symboles** : $n_A, n_B$ = effectifs des groupes ; $\mu_A,\mu_B$ = leurs centres de gravité ; $\|\cdot\|$ = distance euclidienne.
- **Paramètres** : aucun paramètre appris au sens strict (la hiérarchie dépend des données). **Hyperparamètres** : `n_clusters` (ici 3) ; `linkage` (critère de fusion, ici Ward par défaut) et distance (euclidienne par défaut).
- **Avantage 1** : fournit une hiérarchie complète (dendrogramme) qui aide à choisir le nombre de groupes.
- **Avantage 2** : ne dépend pas d'une initialisation aléatoire.
- **Limite 1** : coût de calcul élevé (mémoire et temps) sur de grands jeux de données.
- **Limite 2** : une fusion est définitive (pas de correction ultérieure) ; sensible aux valeurs aberrantes et au choix du critère de fusion.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

rng = np.random.default_rng(0)
X = np.vstack([rng.normal([0, 0], 0.5, (8, 2)),
               rng.normal([4, 4], 0.5, (8, 2)),
               rng.normal([0, 4.5], 0.5, (8, 2))])      # données synthétiques
Z = linkage(X, method="ward")

plt.figure(figsize=(8, 4))
dendrogram(Z, color_threshold=Z[-3, 2] + 0.01, no_labels=True)
plt.axhline(Z[-3, 2] + 0.01, color="red", ls="--", label="Coupure : 3 groupes")
plt.xlabel("Observations"); plt.ylabel("Distance de fusion (Ward)")
plt.title("Classification hiérarchique : dendrogramme (données synthétiques)")
plt.legend(); plt.show()
```

**Lecture du schéma** : l'axe horizontal porte les observations, l'axe vertical la distance à laquelle deux groupes fusionnent. On lit de bas en haut : les points proches fusionnent très bas, les groupes éloignés fusionnent plus haut. La ligne rouge pointillée « coupe » l'arbre : chaque branche qu'elle croise est un groupe final (ici 3, de couleurs différentes). Un grand saut vertical avant une fusion suggère un bon niveau de coupure.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/commerce)** : construire une typologie hiérarchique de produits ou de clients (grandes familles puis sous-familles).
- **Cas d'usage 2 (finance)** : regrouper des actifs ou des pays selon la similarité de leurs rendements ou de leurs indicateurs économiques, et visualiser les rapprochements.
- **Données et prérequis** : variables numériques standardisées, effectif petit à moyen (quelques milliers de points au plus), choix d'une distance et d'un critère de fusion.
- **Cas peu approprié** : très grands jeux de données (matrice de distances trop lourde) ou données bruitées avec beaucoup de points aberrants.
- **Métriques adaptées** : coefficient de silhouette, corrélation cophénétique, lecture du dendrogramme ; validation par l'utilité métier.

### Analyse de l'expérience Python
- **Entrées** : `X_clu` (sans étiquettes). **Hyperparamètres** : `n_clusters=3` ; `linkage` et distance à leurs valeurs par défaut. `fit_predict` renvoie directement les étiquettes (pas d'attribut `labels_` utilisé ici).
- **Sortie** : silhouette = 0,8686, c'est-à-dire la même valeur que pour K-means (0,8686) d'après les deux captures.
- **Ce que cela permet** : constater que, pour $k=3$, ce critère ne départage pas les deux méthodes ; cette coïncidence suggère que les deux algorithmes ont probablement trouvé des regroupements très proches, voire identiques, sur ces données.
- **Ce que cela ne permet pas** : affirmer que les partitions sont identiques (deux partitions différentes pourraient avoir la même silhouette ; il faudrait comparer les étiquettes, par exemple avec l'indice de Rand ajusté) ; prouver que 3 est le bon nombre de groupes ; juger la pertinence métier des groupes ; le code n'affiche pas le dendrogramme.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Présentation
- **Nom** : DBSCAN (clustering spatial basé sur la densité avec bruit) – *Density-Based Spatial Clustering of Applications with Noise*. **Famille** : clustering (apprentissage non supervisé) ; objectif : trouver des groupes comme des zones denses séparées par des zones peu denses, et isoler le bruit.
- **Intuition** : un groupe est une région où les points sont serrés les uns contre les autres ; les points isolés, dans des zones vides, sont déclarés « bruit » plutôt que forcés dans un groupe.
- **Étapes** : pour chaque point, on compte ses voisins dans un rayon `eps` → un point ayant au moins `min_samples` voisins (lui compris) est un point cœur → les points cœur proches s'enchaînent dans un même groupe, auquel s'ajoutent les points frontière → les points restants reçoivent l'étiquette −1 (bruit).
- **Critère** : $N_\varepsilon(x)=\{x' : \|x-x'\|\le\varepsilon\}$ ; $x$ est un point cœur si $|N_\varepsilon(x)|\ge m$.
- **Symboles** : $\varepsilon$ = rayon de voisinage (`eps`) ; $m$ = nombre minimal de points (`min_samples`) ; $N_\varepsilon(x)$ = voisinage de $x$ ; $\|\cdot\|$ = distance euclidienne.
- **Paramètres** : aucun paramètre appris au sens strict (les groupes sont déduits des données). **Hyperparamètres** : `eps` (ici 0,8), `min_samples` (ici 5) ; métrique à sa valeur par défaut.
- **Avantage 1** : ne demande pas le nombre de groupes, trouve des formes quelconques et détecte le bruit.
- **Avantage 2** : robuste aux valeurs aberrantes (elles sont isolées).
- **Limite 1** : très sensible à `eps` et `min_samples` ; difficile quand les groupes ont des densités différentes.
- **Limite 2** : performances dégradées en grande dimension (les distances perdent leur sens) ; pas de prédiction directe pour de nouveaux points.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

X, _ = make_moons(n_samples=200, noise=0.07, random_state=0)   # données synthétiques
bruit = np.random.default_rng(0).uniform(-1.5, 2.5, (10, 2))
X = np.vstack([X, bruit])
lab = DBSCAN(eps=0.2, min_samples=5).fit_predict(X)

plt.figure(figsize=(6, 4.5))
for g in sorted(set(lab)):
    m = lab == g
    if g == -1:
        plt.scatter(*X[m].T, color="black", marker="x", label="Bruit (−1)")
    else:
        plt.scatter(*X[m].T, s=14, label=f"Groupe {g}")
plt.xlabel("Variable 1"); plt.ylabel("Variable 2")
plt.title("DBSCAN : zones denses = groupes, points isolés = bruit (données synthétiques)")
plt.legend(fontsize=8); plt.show()
```

**Lecture du schéma** : les points forment deux demi-lunes, auxquelles s'ajoutent quelques points dispersés. Chaque couleur est un groupe trouvé par densité, même si sa forme est courbe (ce que K-means ne saurait pas faire). Les croix noires sont des points isolés que DBSCAN refuse de rattacher à un groupe : ils reçoivent l'étiquette −1 (bruit).

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : repérer des transactions atypiques isolées par rapport aux comportements habituels (détection d'anomalies simple), les groupes denses représentant l'activité normale.
- **Cas d'usage 2 (logistique/commerce)** : identifier des zones géographiques denses de livraisons ou de clients, et écarter les adresses isolées.
- **Données et prérequis** : variables numériques standardisées, peu de variables, densités de groupes comparables, rayon `eps` choisi (par exemple avec le graphe des distances au $k$-ième voisin).
- **Cas peu approprié** : groupes de densités très différentes, ou grande dimension.
- **Métriques adaptées** : nombre de groupes et proportion de bruit, silhouette (en excluant le bruit, et avec prudence pour les formes non convexes), indice de validité fondé sur la densité ; validation par l'utilité métier.

### Analyse de l'expérience Python
- **Entrées** : `X_clu` (sans étiquettes). **Hyperparamètres** : `eps=0.8`, `min_samples=5`. Le code affiche l'ensemble des étiquettes trouvées et le nombre de points de bruit.
- **Sortie** : groupes {0, 1, 2, −1} et bruit = 8. DBSCAN a donc trouvé 3 groupes (étiquettes 0, 1 et 2) ; l'étiquette −1 désigne le bruit, qui compte 8 points.
- **Ce que cela permet** : constater qu'avec ces réglages, DBSCAN retrouve aussi 3 groupes, comme les deux méthodes précédentes, tout en isolant 8 points jugés atypiques.
- **Ce que cela ne permet pas** : savoir si ces 3 groupes sont les mêmes que ceux de K-means et de la CAH (partitions non comparées) ; connaître la taille des groupes ; juger si `eps=0.8` et `min_samples=5` sont adaptés (valeurs fixées, non validées : un autre `eps` changerait le nombre de groupes et de points de bruit) ; conclure que les 8 points sont de véritables anomalies, car ils sont seulement hors des zones denses.

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

### Présentation
- **Nom** : Modèle de mélange gaussien – *Gaussian Mixture Model (GMM)*. **Famille** : clustering probabiliste (apprentissage non supervisé) ; objectif : modéliser les données comme un mélange de lois normales et affecter chaque point à la composante la plus probable.
- **Intuition** : chaque groupe est une « cloche » gaussienne (centre, forme, taille propres) ; un point appartient à tous les groupes avec une certaine probabilité (affectation « souple »).
- **Étapes** : initialisation des composantes → étape E (probabilité que chaque point provienne de chaque composante) → étape M (mise à jour des poids, moyennes et covariances) → répétition jusqu'à convergence (algorithme EM) → sortie : composante la plus probable.
- **Critère** : $p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal{N}(x\mid\mu_k,\Sigma_k)$, maximisation de la vraisemblance $\sum_i\ln p(x_i)$.
- **Symboles** : $K$ = nombre de composantes ; $\pi_k$ = poids de la composante $k$ ($\sum_k\pi_k=1$) ; $\mu_k$ = moyenne ; $\Sigma_k$ = matrice de covariance ; $\mathcal{N}$ = densité gaussienne.
- **Paramètres** (appris) : $\pi_k$, $\mu_k$, $\Sigma_k$. **Hyperparamètres** : `n_components` (ici 3), `random_state` (ici `GRAINE`) ; type de covariance (`full` par défaut) et autres options par défaut.
- **Avantage 1** : groupes de formes elliptiques et de tailles différentes, avec probabilités d'appartenance.
- **Avantage 2** : modèle génératif, évaluable par critères d'information (BIC, AIC).
- **Limite 1** : $K$ à fixer à l'avance ; dépend de l'initialisation (minima locaux).
- **Limite 2** : suppose des composantes gaussiennes ; nombreux paramètres en grande dimension.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse
from sklearn.mixture import GaussianMixture

rng = np.random.default_rng(0)
X = np.vstack([rng.multivariate_normal([0, 0], [[1.0, 0.8], [0.8, 0.8]], 100),
               rng.multivariate_normal([5, 4], [[0.5, 0], [0, 1.5]], 100),
               rng.multivariate_normal([0, 5], [[1.2, -0.5], [-0.5, 0.6]], 100)])  # données synthétiques
gmm = GaussianMixture(n_components=3, random_state=0).fit(X)
lab = gmm.predict(X)

fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(*X.T, c=lab, cmap="viridis", s=12)
for mu, cov in zip(gmm.means_, gmm.covariances_):
    val, vec = np.linalg.eigh(cov)
    angle = np.degrees(np.arctan2(vec[1, 1], vec[0, 1]))
    for n in (1, 2):   # contours à 1 et 2 écarts-types
        ax.add_patch(Ellipse(mu, 2*n*np.sqrt(val[1]), 2*n*np.sqrt(val[0]), angle=angle,
                             fill=False, ec="red", lw=1.5))
ax.scatter(*gmm.means_.T, color="red", marker="X", s=120, edgecolors="black")
ax.set_xlabel("Variable 1"); ax.set_ylabel("Variable 2")
ax.set_title("Mélange gaussien : 3 ellipses = 3 lois normales (données synthétiques)")
plt.show()
```

**Lecture du schéma** : chaque couleur est une composante du mélange. Les ellipses rouges sont les contours de la loi normale de chaque composante (à 1 et 2 écarts-types) : leur position donne la moyenne (croix), leur orientation et leur allongement donnent la covariance. Contrairement à K-means, les groupes peuvent être allongés et inclinés ; un point situé entre deux ellipses a une probabilité partagée entre les deux composantes.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing)** : segmenter des clients dont les profils se chevauchent, en obtenant la probabilité d'appartenance à chaque segment.
- **Cas d'usage 2 (finance)** : modéliser des régimes de marché (calme, volatil) à partir des rendements, ou repérer des observations de faible vraisemblance (anomalies).
- **Données et prérequis** : variables numériques continues, approximativement gaussiennes dans chaque groupe ; effectif suffisant pour estimer les covariances.
- **Cas peu approprié** : groupes de formes non convexes ou très non gaussiennes (anneaux, demi-lunes), ou faible effectif avec beaucoup de variables.
- **Métriques adaptées** : BIC/AIC pour choisir $K$, log-vraisemblance, silhouette ; validation par l'utilité métier.

### Analyse de l'expérience Python
- **Entrées** : `X_clu` (sans étiquettes). **Hyperparamètres** : `n_components=3`, `random_state=GRAINE` ; autres options par défaut.
- **Sortie** : poids = [0,3333 0,3333 0,3333] ; silhouette = 0,8686.
- **Ce que cela permet** : constater que les trois composantes ont des poids égaux, c'est-à-dire que le modèle attribue un tiers des données à chacune ; et que la silhouette (0,8686) est identique à celles de K-means et de la CAH d'après les captures précédentes : les trois méthodes semblent donc retrouver des groupes très proches.
- **Ce que cela ne permet pas** : affirmer que les partitions sont identiques sans comparer les étiquettes (par exemple avec l'indice de Rand ajusté) ; connaître les moyennes et covariances (non affichées) ; confirmer que $K=3$ est optimal (BIC/AIC non calculés). La silhouette est calculée sur les affectations « dures » et ne mesure pas la qualité probabiliste du modèle.

- **Nom** : Analyse en composantes principales (ACP) – *Principal Component Analysis (PCA)*. **Famille** : réduction de dimension (apprentissage non supervisé) ; objectif : résumer de nombreuses variables corrélées par un petit nombre de nouveaux axes qui conservent le plus de variance.
- **Intuition** : on cherche les directions dans lesquelles les données sont le plus étalées, puis on projette les observations sur les premières d'entre elles.
- **Étapes** : centrage et standardisation → calcul de la matrice de covariance (ou décomposition en valeurs singulières) → axes principaux classés par variance décroissante → projection sur les `n_components` premiers axes.
- **Critère** : le premier axe $w_1=\underset{\|w\|=1}{\arg\max}\ \mathrm{Var}(Xw)$ ; part de variance expliquée par l'axe $j$ : $\dfrac{\lambda_j}{\sum_{l=1}^{p}\lambda_l}$.
- **Symboles** : $X$ = données centrées ; $w_j$ = vecteur directeur de l'axe $j$ (unitaire) ; $\lambda_j$ = valeur propre de la covariance, soit la variance portée par l'axe $j$ ; $p$ = nombre de variables.
- **Paramètres** (appris) : axes $w_j$ et variances $\lambda_j$. **Hyperparamètres** : `n_components` (ici 2) ; prétraitement `StandardScaler`.
- **Avantage 1** : réduit la dimension et supprime la redondance (variables corrélées), ce qui permet de visualiser en 2D.
- **Avantage 2** : méthode rapide, déterministe, sans paramètre délicat.
- **Limite 1** : uniquement linéaire ; ne capte pas des structures courbes.
- **Limite 2** : les axes sont des combinaisons de variables, donc difficiles à interpréter ; sensible à l'échelle (standardisation nécessaire) et aux valeurs aberrantes.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

rng = np.random.default_rng(0)
X = rng.multivariate_normal([0, 0], [[3, 2.2], [2.2, 2]], 150)   # données synthétiques
pca = PCA(n_components=2).fit(X)
mu = X.mean(axis=0)

plt.figure(figsize=(6, 5))
plt.scatter(*X.T, s=12, color="tab:blue", label="Observations")
for i, c in enumerate(["red", "green"]):
    v = pca.components_[i] * 2 * np.sqrt(pca.explained_variance_[i])
    plt.arrow(*mu, *v, color=c, width=0.05, head_width=0.25,
              label=f"Axe {i+1} ({pca.explained_variance_ratio_[i]*100:.0f} % de la variance)")
plt.axis("equal")
plt.xlabel("Variable 1"); plt.ylabel("Variable 2")
plt.title("ACP : axes de plus grande variance (données synthétiques)")
plt.legend(); plt.show()
```

**Lecture du schéma** : le nuage de points est allongé selon une diagonale. La flèche rouge (axe 1) suit la direction où les données sont le plus étalées ; la flèche verte (axe 2), perpendiculaire, porte la variance restante. La légende indique la part de variance de chaque axe. Projeter les points sur l'axe 1 seul résume le nuage en une dimension en perdant peu d'information.

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : résumer des dizaines de ratios ou de rendements corrélés en quelques facteurs (par exemple, facteur de marché), pour visualiser ou alimenter un modèle.
- **Cas d'usage 2 (marketing/commerce)** : visualiser en 2D des clients ou des produits décrits par de nombreuses variables (enquête de satisfaction, caractéristiques) et repérer les proximités.
- **Données et prérequis** : variables numériques continues, standardisées, corrélées entre elles, sans valeurs manquantes.
- **Cas peu approprié** : structure non linéaire (spirale, variétés courbes), ou besoin de conserver des variables directement interprétables.
- **Métriques adaptées** : variance expliquée (cumulée), éboulis des valeurs propres, erreur de reconstruction ; utilité en aval (performance d'un modèle sur les composantes).

### Analyse de l'expérience Python
- **Entrées** : `load_wine().data`, standardisé par `StandardScaler` (`X_vin`) ; pas d'étiquettes ni de découpe train/test. **Hyperparamètres** : `n_components=2`. La sortie comprend la variance expliquée et un nuage de points de la projection.
- **Sortie** : variance expliquée = [0,3620 0,1921], soit environ 36,2 % pour la première composante et 19,2 % pour la seconde, donc environ 55,4 % cumulés.
- **Ce que cela permet** : constater que deux composantes conservent un peu plus de la moitié de l'information (variance) du jeu de données, ce qui rend une visualisation en 2D utile mais incomplète.
- **Ce que cela ne permet pas** : conclure que 2 composantes suffisent (près de 45 % de la variance est perdue) ; connaître le nombre de variables d'origine ou les variables qui pèsent sur chaque axe (loadings non affichés) ; interpréter le nuage de points sans voir le graphique ; en déduire l'existence de groupes (l'ACP ne détecte pas de classes).

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Présentation
- **Nom** : t-SNE (plongement stochastique de voisins avec loi de Student) – *t-distributed Stochastic Neighbor Embedding*. **Famille** : réduction de dimension non linéaire (non supervisée) ; objectif : visualiser en 2D des données de grande dimension en préservant les voisinages.
- **Intuition** : deux points proches dans l'espace d'origine doivent rester proches dans la carte 2D, et les points éloignés doivent rester éloignés.
- **Étapes** : calcul de probabilités de voisinage entre points dans l'espace d'origine → placement aléatoire des points en 2D → déplacement des points par descente de gradient pour que les voisinages 2D ressemblent à ceux d'origine → sortie : coordonnées 2D.
- **Critère** : minimiser $\mathrm{KL}(P\,\|\,Q)=\sum_{i\ne j}p_{ij}\ln\dfrac{p_{ij}}{q_{ij}}$, avec $q_{ij}\propto\left(1+\|y_i-y_j\|^2\right)^{-1}$.
- **Symboles** : $p_{ij}$ = similarité de $i$ et $j$ dans l'espace d'origine ; $q_{ij}$ = similarité dans la carte ; $y_i$ = position 2D du point $i$ ; KL = divergence de Kullback-Leibler.
- **Paramètres** (appris) : coordonnées $y_i$ de chaque point. **Hyperparamètres** : `n_components` (ici 2), `perplexity` (ici 20, taille approximative du voisinage), `random_state` (ici `GRAINE`).
- **Avantage 1** : révèle très bien des groupes locaux et des structures non linéaires.
- **Avantage 2** : visualisations souvent plus lisibles que l'ACP pour des données complexes.
- **Limite 1** : les distances entre groupes et la taille des groupes ne sont pas interprétables ; le résultat dépend de `perplexity` et de l'initialisation.
- **Limite 2** : coûteux sur de grands jeux de données et sans transformation directe de nouveaux points.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

rng = np.random.default_rng(0)
centres = rng.normal(0, 4, (3, 10))                       # 3 groupes dans un espace à 10 dimensions
X = np.vstack([c + rng.normal(0, 1.5, (60, 10)) for c in centres])   # données synthétiques
etiq = np.repeat([0, 1, 2], 60)

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
for ax, proj, titre in [(axes[0], PCA(n_components=2).fit_transform(X), "ACP (projection linéaire)"),
                        (axes[1], TSNE(n_components=2, perplexity=20, random_state=0).fit_transform(X),
                         "t-SNE (perplexity = 20)")]:
    ax.scatter(*proj.T, c=etiq, cmap="viridis", s=14)
    ax.set_title(titre); ax.set_xlabel("Axe 1"); ax.set_ylabel("Axe 2")
fig.suptitle("Même jeu de données en 10 dimensions projeté en 2D (données synthétiques ; couleurs = groupes réels)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : les deux graphiques représentent les mêmes 180 observations en 10 dimensions, projetées en 2D. La couleur indique le groupe réel (connu ici car les données sont synthétiques). À gauche, l'ACP projette de façon linéaire ; à droite, t-SNE reconstruit une carte qui conserve les voisinages. Seule la proximité locale est à lire : la distance entre deux groupes éloignés sur la carte n'a pas de sens précis.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing)** : explorer visuellement une base de clients décrits par de nombreuses variables pour repérer des segments naturels.
- **Cas d'usage 2 (finance/logistique)** : visualiser des transactions, des produits ou des profils de risque pour repérer des regroupements ou des cas atypiques.
- **Données et prérequis** : variables numériques standardisées, effectif petit à moyen (quelques milliers de points), idéalement après une première réduction par ACP.
- **Cas peu approprié** : tout usage qui exige de mesurer des distances ou des tailles de groupes, de transformer de nouveaux points, ou de servir d'entrée à un modèle de production.
- **Métriques adaptées** : divergence KL finale, indice de confiance (*trustworthiness*) et préservation des voisins ; validation visuelle en comparant plusieurs valeurs de `perplexity`.

### Analyse de l'expérience Python
- **Entrées** : `load_wine().data` standardisé par `StandardScaler` (`X_vin`) ; pas d'étiquettes ni de découpe train/test. **Hyperparamètres** : `n_components=2`, `perplexity=20`, `random_state=GRAINE`.
- **Sortie** : aucune métrique n'est affichée ; le code produit uniquement un nuage de points 2D intitulé « t-SNE ».
- **Ce que cela permet** : visualiser si les observations se regroupent en zones distinctes une fois projetées en 2D (à examiner sur votre graphique complet).
- **Ce que cela ne permet pas** : mesurer la qualité du plongement (aucune métrique) ; interpréter les distances entre groupes ou leur taille ; affirmer qu'il existe un nombre précis de groupes ; vérifier la stabilité du résultat face à d'autres valeurs de `perplexity` ou d'autres graines (non testées).

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
```

### Présentation
- **Nom** : UMAP (approximation et projection uniformes de variétés) – *Uniform Manifold Approximation and Projection*. **Famille** : réduction de dimension non linéaire (non supervisée) ; objectif : projeter des données en basse dimension en préservant la structure locale et en partie la structure globale.
- **Intuition** : on suppose que les données se trouvent sur une surface courbe (variété) de dimension plus petite ; on construit un graphe de voisinage, puis on dessine un graphe 2D qui lui ressemble.
- **Étapes** : recherche des `n_neighbors` plus proches voisins de chaque point → construction d'un graphe pondéré (« flou ») dans l'espace d'origine → initialisation d'un graphe en 2D → optimisation des positions pour que les deux graphes se ressemblent → sortie : coordonnées 2D.
- **Critère** : minimiser l'entropie croisée $\sum_{(i,j)}\left[\mu_{ij}\ln\dfrac{\mu_{ij}}{\nu_{ij}}+(1-\mu_{ij})\ln\dfrac{1-\mu_{ij}}{1-\nu_{ij}}\right]$.
- **Symboles** : $\mu_{ij}$ = poids de l'arête $i$–$j$ dans l'espace d'origine ; $\nu_{ij}$ = poids correspondant dans la carte 2D.
- **Paramètres** (appris) : coordonnées de chaque point en 2D. **Hyperparamètres** : `n_components` (ici 2), `n_neighbors` (ici 15, taille du voisinage), `random_state` (ici `GRAINE`) ; `min_dist` et la métrique à leurs valeurs par défaut.
- **Avantage 1** : plus rapide que t-SNE et conserve mieux la structure globale.
- **Avantage 2** : peut transformer de nouveaux points une fois le modèle ajusté.
- **Limite 1** : sensible à `n_neighbors` et à `min_dist` ; les distances restent approximatives.
- **Limite 2** : bibliothèque externe (`umap-learn`) ; fixer `random_state` fige la graine mais désactive le calcul parallèle.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
from sklearn.neighbors import NearestNeighbors

rng = np.random.default_rng(0)
X = np.vstack([rng.normal([0, 0], 0.5, (12, 2)), rng.normal([3, 2], 0.5, (12, 2))])   # données synthétiques
voisins = NearestNeighbors(n_neighbors=3).fit(X).kneighbors(X, return_distance=False)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4), gridspec_kw={"width_ratios": [1, 1.6]})
# Gauche : graphe de voisinage
for i, ligne in enumerate(voisins):
    for j in ligne[1:]:
        ax1.plot(*X[[i, j]].T, color="gray", lw=0.8)
ax1.scatter(*X.T, color="tab:blue", zorder=3)
ax1.set_title("Étape 1 : graphe des voisins (k = 3 ici)")
ax1.set_xlabel("Variable 1"); ax1.set_ylabel("Variable 2")
# Droite : étapes
ax2.set_xlim(0, 10); ax2.set_ylim(0, 4); ax2.axis("off")
etapes = ["Données\nhaute dim.", "Graphe pondéré\n(n_neighbors)", "Initialisation\n2D", "Optimisation\n(entropie croisée)", "Carte 2D"]
for k, txt in enumerate(etapes):
    x = 0.1 + k * 2.0
    ax2.add_patch(FancyBboxPatch((x, 1.4), 1.6, 1.2, boxstyle="round,pad=0.05", fc="#dbe9f6", ec="black"))
    ax2.text(x + 0.8, 2.0, txt, ha="center", va="center", fontsize=8)
    if k < 4:
        ax2.annotate("", xy=(x + 2.0, 2.0), xytext=(x + 1.7, 2.0), arrowprops=dict(arrowstyle="->", lw=1.5))
ax2.set_title("Principe de UMAP (schéma conceptuel)")
plt.tight_layout(); plt.show()
```

**Lecture du schéma** : à gauche, chaque point est relié à ses plus proches voisins : c'est le graphe de voisinage, point de départ de UMAP (ici avec 3 voisins pour la lisibilité). À droite, la chaîne d'étapes se lit de gauche à droite : le graphe de l'espace d'origine est comparé à un graphe en 2D, dont les positions sont ajustées jusqu'à ce que les deux se ressemblent, ce qui donne la carte finale.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/commerce)** : cartographier de grandes bases de clients ou de produits décrits par de nombreux attributs et repérer des regroupements.
- **Cas d'usage 2 (finance/logistique)** : visualiser des profils de transactions ou de trajets, ou réduire la dimension avant un clustering.
- **Données et prérequis** : variables numériques standardisées, effectif moyen à grand ; installation de `umap-learn`.
- **Cas peu approprié** : besoin d'axes interprétables (l'ACP est préférable), ou de mesures précises de distances et de tailles de groupes.
- **Métriques adaptées** : *trustworthiness*, préservation des voisins, stabilité selon `n_neighbors` et `min_dist` ; utilité en aval (par exemple qualité d'un clustering sur la projection).

### Analyse de l'expérience Python
- **Entrées** : `load_wine().data` standardisé (`X_vin`) ; pas d'étiquettes ni de découpe train/test. **Hyperparamètres** : `n_components=2`, `n_neighbors=15`, `random_state=GRAINE` ; `min_dist` et la métrique par défaut.
- **Sortie** : un avertissement (*UserWarning*) puis un nuage de points 2D intitulé « UMAP ». L'avertissement indique que fixer `random_state` impose `n_jobs=1` : le calcul est reproductible mais non parallèle. Ce n'est pas une erreur.
- **Ce que cela permet** : obtenir une carte 2D reproductible et examiner si les observations forment des zones séparées (à examiner sur votre graphique complet).
- **Ce que cela ne permet pas** : mesurer la qualité de la projection (aucune métrique) ; interpréter les distances et les tailles de groupes ; conclure à un nombre précis de groupes ; vérifier la stabilité face à d'autres valeurs de `n_neighbors` ou d'autres graines (non testées).

```python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
```

### Présentation
- **Nom** : Forêt d'isolation – *Isolation Forest*. **Famille** : détection d'anomalies (non supervisée) ; objectif : repérer les observations atypiques sans étiquettes.
- **Intuition** : un point atypique est isolé du reste par très peu de coupures aléatoires, alors qu'un point normal, noyé dans la masse, en demande beaucoup.
- **Étapes** : construction de nombreux arbres où chaque coupure choisit une variable et un seuil au hasard → mesure de la profondeur à laquelle chaque point est isolé → moyenne sur tous les arbres → score d'anomalie → les points les plus anormaux (selon `contamination`) sont étiquetés −1.
- **Critère** : $s(x,n)=2^{-\,E[h(x)]/c(n)}$, proche de 1 pour une anomalie, nettement inférieur à 0,5 pour un point normal.
- **Symboles** : $h(x)$ = profondeur d'isolement de $x$ dans un arbre ; $E[h(x)]$ = moyenne sur la forêt ; $n$ = taille de l'échantillon ; $c(n)$ = profondeur moyenne attendue (terme de normalisation).
- **Paramètres** (appris) : structure des arbres aléatoires. **Hyperparamètres** : `contamination` (ici 0,02, proportion attendue d'anomalies), `random_state` (ici `GRAINE`) ; `n_estimators` (100 par défaut) et autres options par défaut.
- **Avantage 1** : rapide, adapté aux grands jeux de données et à de nombreuses variables, sans standardisation.
- **Avantage 2** : ne suppose aucune distribution et n'a pas besoin d'étiquettes.
- **Limite 1** : `contamination` impose un nombre d'anomalies : le seuil est donc en partie arbitraire.
- **Limite 2** : moins efficace pour des anomalies cachées au milieu des données normales ou en très haute dimension ; résultat non explicatif (pas de raison fournie).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

rng = np.random.default_rng(0)
X = np.vstack([rng.normal([0, 0], 1.0, (200, 2)), [[7, 7], [-6, 6], [6, -7]]])   # données synthétiques
modele = IsolationForest(contamination=0.02, random_state=0).fit(X)
score = modele.decision_function(X)           # plus petit = plus atypique
est_anom = modele.predict(X) == -1

xx, yy = np.meshgrid(np.linspace(-9, 9, 200), np.linspace(-9, 9, 200))
zz = modele.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6.5, 5))
plt.contourf(xx, yy, zz, levels=15, cmap="Blues_r", alpha=0.6)
plt.colorbar(label="Score de normalité (bas = atypique)")
plt.scatter(*X[~est_anom].T, color="tab:blue", s=12, label="Normal")
plt.scatter(*X[est_anom].T, color="red", marker="x", s=60, label="Atypique (−1)")
plt.xlabel("Variable 1"); plt.ylabel("Variable 2")
plt.title("Isolation Forest : les points isolés ont un score bas (données synthétiques)")
plt.legend(); plt.show()
```

**Lecture du schéma** : le fond coloré est le score de normalité appris par le modèle : foncé au centre du nuage (zone dense, normale), de plus en plus clair vers l'extérieur (zones où un point serait vite isolé). Les points bleus sont jugés normaux ; les croix rouges, situées loin du nuage, ont un score bas et reçoivent l'étiquette −1 (atypique).

### Contextes d'utilisation
- **Cas d'usage 1 (finance)** : détecter des transactions bancaires ou des écritures comptables suspectes (fraude, erreurs de saisie).
- **Cas d'usage 2 (logistique/commerce international)** : repérer des commandes, des déclarations douanières ou des délais de livraison aberrants.

- **Cas peu approprié** : anomalies qui imitent le comportement normal (fraude sophistiquée) ou fraction d'anomalies très importante ; contexte où chaque alerte doit être justifiée précisément.
- **Métriques adaptées** : sans étiquettes, nombre d'alertes et revue humaine d'un échantillon ; avec quelques étiquettes, précision, rappel, F1, AUC-ROC, précision@k.

### Analyse de l'expérience Python

- **Sortie** : observations atypiques (indices) = [20, 79, 130, 147, 240], soit 5 observations marquées −1.

Les points injectés se trouvent en fin de tableau : si `X_clu` compte 240 observations, l'indice 240 correspond à [9, 9] et l'indice 241 à [−8, −8]. Dans ce cas, un seul des deux points injectés est détecté ; le second (241) ne l'est pas. Vérifiez `X_clu.shape[0]`.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Présentation
- **Nom** : Bandits manchots (à plusieurs bras) – *Multi-armed Bandits*. **Famille** : apprentissage par renforcement (cas à un seul état, sans transition) ; objectif : choisir entre plusieurs actions incertaines pour maximiser la récompense cumulée.
- **Intuition** : on dispose de plusieurs « machines » dont on ignore le taux de gain ; il faut explorer pour les connaître et exploiter la meilleure pour gagner.
- **Étapes** : à chaque pas, choix d'un bras → observation d'une récompense (gain 0 ou 1) → mise à jour de l'estimation de ce bras. Dans votre code, le choix est cyclique (`bras = t % 3`), sans stratégie.
- **Critère** : estimation $\hat{\mu}_a=\dfrac{S_a}{N_a}$ ; objectif de minimiser le regret $R_T=T\mu^*-\sum_{t=1}^{T}\mu_{a_t}$.
- **Symboles** : $a$ = bras ; $S_a$ = somme des gains du bras $a$ ; $N_a$ = nombre de tirages du bras $a$ ; $\mu_a$ = vraie probabilité de gain ; $\mu^*$ = meilleure probabilité ; $T$ = nombre de pas.
- **Paramètres** (appris) : estimations $\hat{\mu}_a$. **Hyperparamètres** : nombre de pas (ici 500), vraies probabilités simulées [0,2 ; 0,5 ; 0,7] (inconnues du modèle en situation réelle), graine `GRAINE` ; une vraie stratégie ajoute son propre paramètre ($\varepsilon$ pour $\varepsilon$-greedy, par exemple).
- **Avantage 1** : simple, apprend en ligne, sans modèle de l'environnement.
- **Avantage 2** : arbitre explicitement l'exploration et l'exploitation, ce qui réduit le coût d'apprentissage par rapport à un test A/B fixe.
- **Limite 1** : suppose que les gains de chaque bras sont stables dans le temps et que l'action n'influence pas l'état futur.
- **Limite 2** : le choix de la stratégie (ε-greedy, UCB, Thompson) change fortement le résultat, et les récompenses doivent être observées rapidement.

```python
import numpy as np
import matplotlib.pyplot as plt

g = np.random.default_rng(0)
vraies = np.array([0.2, 0.5, 0.7])
somme = np.zeros(3); comptes = np.zeros(3)
historique = [[], [], []]
for t in range(500):                      # même principe que le code fourni (simulation, graine arbitraire)
    bras = t % 3
    somme[bras] += int(g.random() < vraies[bras]); comptes[bras] += 1
    for a in range(3):
        historique[a].append(somme[a] / comptes[a] if comptes[a] else np.nan)

fig, ax = plt.subplots(figsize=(7, 4))
for a, c in enumerate(["tab:blue", "tab:orange", "tab:green"]):
    ax.plot(historique[a], color=c, label=f"Estimation du bras {a}")
    ax.axhline(vraies[a], color=c, ls="--", lw=0.8)
ax.plot([], [], color="gray", ls="--", label="Vraie probabilité de gain")
ax.set_xlabel("Pas de temps t"); ax.set_ylabel("Gain moyen estimé")
ax.set_title("Bandits : l'estimation de chaque bras converge vers sa vraie valeur (simulation)")
ax.legend(); plt.show()
```

**Lecture du schéma** : chaque courbe pleine est l'estimation du gain moyen d'un bras, mise à jour au fil des tirages ; la ligne pointillée de même couleur est sa vraie probabilité de gain, inconnue du modèle. Au début, les courbes fluctuent beaucoup (peu de tirages) ; plus il y a de tirages, plus elles se rapprochent de leur valeur vraie. Ce schéma suit le même tirage cyclique que votre code, c'est-à-dire sans exploitation : une vraie stratégie de bandit tirerait de plus en plus souvent le meilleur bras.

### Contextes d'utilisation

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Présentation
- **Nom** : Stratégie ε-greedy (ε-glouton) pour bandits manchots – *Epsilon-greedy*. **Famille** : apprentissage par renforcement (cas à un seul état) ; objectif : maximiser la récompense cumulée en arbitrant entre exploration et exploitation.
- **Intuition** : la plupart du temps on joue le bras qui paraît le meilleur (exploitation) ; avec une petite probabilité ε, on joue un bras au hasard pour continuer à en apprendre davantage (exploration).
- **Étapes** : avec probabilité ε → bras tiré au hasard ; sinon → bras d'estimation maximale → observation du gain (0 ou 1) → mise à jour de la somme des gains et du nombre de tirages du bras.
- **Critère** : $a_t=\begin{cases}\text{bras au hasard} & \text{avec probabilité }\varepsilon\\ \arg\max_a\ \hat{\mu}_a & \text{sinon}\end{cases}$, avec $\hat{\mu}_a=\dfrac{S_a+1}{N_a+1}$ dans votre code.
- **Symboles** : $a$ = bras ; $\hat{\mu}_a$ = valeur estimée du bras $a$ ; $S_a$ = somme de ses gains ; $N_a$ = nombre de fois où il a été joué ; $\varepsilon$ = probabilité d'explorer.
- **Paramètres** (appris) : les estimations $\hat{\mu}_a$ (via `somme` et `comptes`). **Hyperparamètres** : `epsilon` (ici 0,1), nombre de pas (ici 1000), graine `GRAINE`.
- **Particularité du code** : le « +1 » au numérateur et au dénominateur donne la valeur 1 à un bras jamais joué, ce qui pousse à tester chaque bras au début.
- **Avantage 1** : très simple à implémenter, sans modèle ni calcul lourd.
- **Avantage 2** : garantit que tous les bras continuent d'être testés.
- **Limite 1** : explore toujours au hasard, même quand un bras est clairement mauvais (ε fixe = perte permanente).
- **Limite 2** : suppose des gains stables dans le temps ; le choix de ε est délicat.

```python
import numpy as np
import matplotlib.pyplot as plt

g = np.random.default_rng(0)
vraies = np.array([0.2, 0.5, 0.7]); epsilon = 0.1
somme = np.zeros(3); comptes = np.zeros(3)
hist = np.zeros((1000, 3))
for t in range(1000):                     # même règle que le code fourni (simulation, graine arbitraire)
    bras = int(g.integers(3)) if g.random() < epsilon else int(np.argmax((somme + 1) / (comptes + 1)))
    somme[bras] += int(g.random() < vraies[bras]); comptes[bras] += 1
    hist[t] = comptes

plt.figure(figsize=(7, 4))
for a, c in enumerate(["tab:blue", "tab:orange", "tab:green"]):
    plt.plot(hist[:, a], color=c, label=f"Bras {a} (vraie proba = {vraies[a]})")
plt.xlabel("Pas de temps t"); plt.ylabel("Nombre cumulé de choix du bras")
plt.title("ε-greedy : le meilleur bras est choisi de plus en plus souvent (simulation)")
plt.legend(); plt.show()
```

**Lecture du schéma** : chaque courbe donne le nombre cumulé de fois où un bras a été choisi. Au début, les courbes montent à des rythmes proches (tous les bras sont testés). Puis la courbe du meilleur bras (vert) s'accélère nettement, car l'algorithme l'exploite. Les deux autres bras continuent de monter très lentement : c'est la part d'exploration (ε).

### Contextes d'utilisation
- **Cas d'usage 1 (marketing)** : choisir en continu entre plusieurs versions d'une bannière, d'un objet d'e-mail ou d'une page produit pour maximiser le taux de clic ou d'achat.
- **Cas d'usage 2 (commerce/finance)** : tester plusieurs prix ou plusieurs offres promotionnelles, ou répartir des ordres entre plateformes d'exécution selon leur performance observée.
- **Données et prérequis** : un petit nombre d'actions, une récompense mesurable rapidement après chaque action, un environnement raisonnablement stable.
- **Cas peu approprié** : quand l'action modifie l'état futur (gestion de stocks, trading dynamique), quand les gains évoluent fortement dans le temps, ou quand une erreur coûte très cher (l'exploration aléatoire est alors risquée).
- **Métriques adaptées** : regret cumulé, récompense moyenne, part des choix portant sur le meilleur bras, erreur d'estimation $|\hat{\mu}_a-\mu_a|$ (en simulation).

### Analyse de l'expérience Python
- **Entrées** : aucune donnée externe ; simulation de 1000 tirages sur 3 bras de vraies probabilités [0,2 ; 0,5 ; 0,7]. **Hyperparamètres** : `epsilon=0.1`, 1000 pas, graine `GRAINE`.
- **Sortie** : nombre de choix = [25, 49, 926] ; valeurs estimées = [0,2 ; 0,6735 ; 0,6933].
- **Ce que cela permet** : constater que le meilleur bras (le 3ᵉ, de vraie probabilité 0,7) est joué 926 fois sur 1000, soit 92,6 % des tirages, et que son estimation (0,693) est proche de sa vraie valeur. Avec les vraies probabilités du code, la récompense espérée de ces choix est d'environ 677,7 contre 700 en jouant toujours le meilleur bras, soit un regret espéré d'environ 22 (calcul à partir des comptes affichés, non produit par le code).
- **Ce que cela ne permet pas** : l'estimation du bras 1 (0,673) est loin de sa vraie valeur (0,5), car il n'a été joué que 49 fois : elle est bruitée, et cette proximité avec celle du bras 2 peut expliquer pourquoi il est encore tiré. Les estimations affichées (`somme/max(comptes,1)`) diffèrent de celles qui guident le choix (`(somme+1)/(comptes+1)`). Une seule simulation ne dit rien de la variabilité selon la graine ; le code ne calcule pas le regret ; on ne sait pas si un autre ε ferait mieux (non testé).

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Présentation
- **Nom** : Q-learning (apprentissage par différence temporelle, hors politique) – *Q-learning*. **Famille** : apprentissage par renforcement ; objectif : apprendre une politique (quelle action jouer dans chaque état) qui maximise la récompense cumulée.
- **Intuition** : un agent explore un environnement et mémorise dans une table $Q$ la valeur de chaque couple (état, action), c'est-à-dire la récompense future attendue ; il en déduit la meilleure action à jouer.
- **Étapes** : état courant → choix d'une action (ε-greedy) → observation de la récompense et du nouvel état → mise à jour de $Q$ avec la meilleure valeur du nouvel état → répétition sur de nombreux épisodes → politique = action de $Q$ maximale dans chaque état.
- **Critère** : $Q(s,a)\leftarrow Q(s,a)+\alpha\left[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\right]$
- **Symboles** : $s,s'$ = état courant et suivant ; $a$ = action ; $r$ = récompense ; $\alpha$ = taux d'apprentissage ; $\gamma$ = facteur d'actualisation (poids du futur) ; $Q(s,a)$ = valeur de l'action $a$ dans l'état $s$.
- **Paramètres** (appris) : la table $Q$ (ici 16 états × 4 actions = 64 valeurs). **Hyperparamètres** : $\alpha=0{,}3$, $\gamma=0{,}95$, 4000 épisodes, 100 pas maximum par épisode, $\varepsilon=\max(0{,}05;\,1-\text{épisode}/2500)$ (exploration forte au début, qui tombe à 5 % vers l'épisode 2375).
- **Avantage 1** : algorithme simple, sans modèle de l'environnement ; converge vers la politique optimale sous conditions (exploration suffisante, espace d'états discret).
- **Avantage 2** : « hors politique » : il apprend la meilleure politique même en explorant.
- **Limite 1** : table $Q$ impraticable quand l'espace d'états est grand ou continu (d'où le DQN).
- **Limite 2** : sensible à $\alpha$, $\gamma$ et au calendrier d'exploration ; peut être trop optimiste quand l'exploration est risquée.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(9, 4.8))
ax.set_xlim(0, 10); ax.set_ylim(0, 6); ax.axis("off")

def boite(x, y, w, h, txt, c):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.1", fc=c, ec="black"))
    ax.text(x + w/2, y + h/2, txt, ha="center", va="center", fontsize=9)

boite(0.3, 3.5, 3.0, 1.5, "Agent\nTable Q[état, action]", "#fde9c9")
boite(6.7, 3.5, 3.0, 1.5, "Environnement\n(FrozenLake 4×4)", "#dbe9f6")
ax.annotate("", xy=(6.6, 4.5), xytext=(3.4, 4.5), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(5.0, 4.7, "action a (ε-greedy)", ha="center", fontsize=9)
ax.annotate("", xy=(3.4, 3.9), xytext=(6.6, 3.9), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(5.0, 3.4, "nouvel état s', récompense r", ha="center", fontsize=9)
boite(1.0, 0.6, 8.0, 1.5, "Mise à jour : Q(s,a) ← Q(s,a) + α [ r + γ · max Q(s',·) − Q(s,a) ]", "#d7f0d2")
ax.annotate("", xy=(1.8, 2.3), xytext=(1.8, 3.4), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(2.0, 2.85, "met à jour", fontsize=8)
ax.set_title("Q-learning : boucle agent–environnement et mise à jour de la table Q (schéma conceptuel)")
plt.show()
```

**Lecture du schéma** : en haut, la boucle d'interaction. L'agent (qui possède la table Q) envoie une action à l'environnement, qui renvoie un nouvel état et une récompense. En bas, la formule met à jour la table : la valeur de l'action jouée est rapprochée de « récompense obtenue + valeur de la meilleure action possible depuis le nouvel état, pondérée par γ ». Répétée des milliers de fois, cette boucle propage la récompense du but vers les états de départ.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique)** : apprendre un itinéraire ou une stratégie de routage dans un réseau discret (entrepôt, graphe de livraison) avec un nombre limité d'états.
- **Cas d'usage 2 (gestion/commerce)** : politique de réapprovisionnement ou de tarification avec des niveaux de stock et de prix discrétisés (en simulation d'abord).
- **Données et prérequis** : un environnement simulable ou interactif, états et actions discrets en petit nombre, une fonction de récompense bien définie.
- **Cas peu approprié** : espaces d'états grands ou continus (table ingérable), ou situations où explorer dans la réalité est coûteux ou dangereux (il faut alors un simulateur).
- **Métriques adaptées** : récompense cumulée par épisode, taux de réussite, longueur moyenne des trajets, stabilité de la politique entre graines ; évaluation sur des épisodes séparés de l'entraînement.

### Analyse de l'expérience Python
- **Entrées** : environnement `FrozenLake-v1` (16 états, 4 actions), `is_slippery=False` (mouvements déterministes) ; table `Q` de zéros ; graine `GRAINE+episode` à chaque réinitialisation. **Hyperparamètres** : 4000 épisodes, 100 pas maximum, $\alpha=0{,}3$, $\gamma=0{,}95$, $\varepsilon=\max(0{,}05;\,1-\text{épisode}/2500)$.
- **Sortie** : politique = [1 2 1 0 1 0 1 0 2 1 1 0 0 2 2 0], soit l'action choisie (0 = gauche, 1 = bas, 2 = droite, 3 = haut) pour chacun des 16 états (numérotés ligne par ligne).
- **Ce que cela permet** : sur la carte 4×4 par défaut, la lecture de cette politique depuis l'état 0 donne le chemin bas, bas, droite, bas, droite, droite, qui atteint l'état 15 (but) en 6 pas sans tomber dans un trou. La politique apprise semble donc correcte pour aller de l'état de départ au but.
- **Ce que cela ne permet pas** : les états-trous (5, 7, 11, 12) et le but (15) n'ont aucune mise à jour de `Q` : leur action « 0 » vient de `argmax` sur des zéros et n'a pas de sens. Le code ne simule pas la politique et n'affiche ni récompense ni taux de réussite : ce chemin est déduit de la lecture. On ne peut rien dire de l'optimalité sur d'autres environnements, ni sur la vitesse de convergence (aucune courbe d'apprentissage).

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Présentation
- **Nom** : SARSA (État–Action–Récompense–État–Action, apprentissage sur la politique) – *SARSA (State–Action–Reward–State–Action)*. **Famille** : apprentissage par renforcement ; objectif : apprendre la valeur des actions en tenant compte de la politique réellement suivie, y compris ses actions exploratoires.
- **Intuition** : comme Q-learning, mais la mise à jour s'appuie sur l'action que l'agent va réellement jouer ensuite, et non sur la meilleure action possible : il apprend donc une politique « prudente », cohérente avec sa propre exploration.
- **Étapes** : état $s$ et action $a$ → observation de $r$ et $s'$ → choix de l'action suivante $a'$ (ε-greedy) → mise à jour de $Q(s,a)$ avec $Q(s',a')$ → on passe à $(s',a')$ → répétition sur de nombreux épisodes.
- **Critère** : $Q(s,a)\leftarrow Q(s,a)+\alpha\left[r+\gamma\,Q(s',a')-Q(s,a)\right]$
- **Symboles** : $s,s'$ = états courant et suivant ; $a,a'$ = actions courante et suivante ; $r$ = récompense ; $\alpha$ = taux d'apprentissage ; $\gamma$ = facteur d'actualisation.
- **Paramètres** (appris) : la table $Q$ (16 × 4 valeurs). **Hyperparamètres** : $\alpha=0{,}3$, $\gamma=0{,}95$, 4000 épisodes, 100 pas maximum, $\varepsilon=\max(0{,}05;\,1-\text{épisode}/2500)$, graine `GRAINE`.
- **Avantage 1** : apprend en tenant compte du risque lié à l'exploration (utile quand une erreur coûte cher).
- **Avantage 2** : simple, mêmes outils que Q-learning (table, ε-greedy).
- **Limite 1** : politique sur la politique : elle converge vers la politique optimale seulement si ε tend vers 0.
- **Limite 2** : même limite que Q-learning pour les grands espaces d'états (table ingérable) ; sensible à $\alpha$, $\gamma$ et à ε.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(10, 4.8))
ax.set_xlim(0, 10); ax.set_ylim(0, 6); ax.axis("off")

def boite(x, y, w, h, txt, c):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.1", fc=c, ec="black"))
    ax.text(x + w/2, y + h/2, txt, ha="center", va="center", fontsize=9)

# Chaîne de la transition
noeuds = [("état s", 0.3), ("action a", 2.6), ("récompense r\nétat s'", 4.9), ("action a'\n(choisie ε-greedy)", 7.4)]
for txt, x in noeuds:
    boite(x, 4.0, 2.0, 1.2, txt, "#dbe9f6")
for x in [2.4, 4.7, 7.0]:
    ax.annotate("", xy=(x + 0.3, 4.6), xytext=(x, 4.6), arrowprops=dict(arrowstyle="->", lw=1.5))
# Deux cibles
boite(0.3, 0.6, 4.6, 1.6, "Q-learning (hors politique)\ncible = r + γ · max Q(s', ·)\n(meilleure action possible)", "#fde9c9")
boite(5.3, 0.6, 4.5, 1.6, "SARSA (sur la politique)\ncible = r + γ · Q(s', a')\n(action réellement choisie)", "#d7f0d2")
ax.annotate("", xy=(2.6, 2.4), xytext=(5.6, 3.9), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate("", xy=(8.4, 2.4), xytext=(8.4, 3.9), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.set_title("SARSA : la cible utilise l'action a' réellement jouée (schéma conceptuel)")
plt.show()
```

**Lecture du schéma** : la rangée du haut est une transition, lue de gauche à droite : l'agent est dans l'état s, joue l'action a, reçoit la récompense r et arrive en s', puis choisit déjà l'action suivante a'. En bas, on compare les deux cibles de mise à jour. Q-learning (orange) prend la meilleure valeur possible en s' ; SARSA (vert) prend la valeur de l'action a' que l'agent va réellement jouer, exploration comprise. C'est cette seule différence qui rend SARSA plus prudent.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique/robotique)** : navigation d'un chariot ou d'un robot d'entrepôt près d'obstacles ou de zones dangereuses, où les écarts exploratoires sont risqués.
- **Cas d'usage 2 (gestion/finance)** : stratégie de décisions séquentielles (réapprovisionnement, allocation) où l'on veut une politique robuste aux actions d'exploration, évaluée d'abord en simulation.
- **Données et prérequis** : un environnement interactif ou simulé, états et actions discrets en petit nombre, une fonction de récompense bien définie.
- **Cas peu approprié** : espaces d'états grands ou continus (table ingérable), ou cas où seule la politique optimale compte et non la sécurité de l'apprentissage (Q-learning est alors plus direct).
- **Métriques adaptées** : récompense cumulée par épisode, taux de réussite, fréquence des échecs pendant l'apprentissage, stabilité entre graines ; évaluation sur des épisodes séparés.

### Analyse de l'expérience Python
- **Entrées** : `FrozenLake-v1` (`is_slippery=False`), table `Q` de zéros (16 × 4), fonction `choisir` (ε-greedy). **Hyperparamètres** : 4000 épisodes, 100 pas maximum, $\alpha=0{,}3$, $\gamma=0{,}95$, $\varepsilon=\max(0{,}05;\,1-\text{épisode}/2500)$.
- **Sortie** : politique SARSA = [1 0 3 0 1 0 1 0 2 1 1 0 0 2 2 0].
- **Ce que cela permet** : comparer avec Q-learning ([1 2 1 0 1 0 1 0 2 1 1 0 0 2 2 0]) : les deux politiques sont identiques sauf aux états 1 et 2. Depuis l'état 0, la politique SARSA suit le même chemin que Q-learning (bas, bas, droite, bas, droite, droite) et atteint le but.

### Présentation
- **Nom** : Réseau Q profond – *Deep Q-Network (DQN)*. **Famille** : apprentissage par renforcement profond ; objectif : apprendre la valeur des actions avec un réseau de neurones quand l'espace d'états est trop grand pour une table $Q$.
- **Intuition** : on remplace la table Q de Q-learning par un réseau de neurones qui prédit, pour un état, la valeur de chaque action ; on l'entraîne à se rapprocher de cibles issues de l'équation de Bellman.
- **Étapes** : l'agent joue (ε-greedy) et stocke ses transitions dans une mémoire de rejeu → tirage d'un mini-lot de transitions → calcul de la cible de Bellman avec un réseau cible figé → descente de gradient sur le réseau en ligne → copie périodique du réseau en ligne vers le réseau cible.
- **Critère** : $y=r+\gamma\,(1-d)\,\max_{a'}Q_{\theta^-}(s',a')$ ; perte $\left(y-Q_\theta(s,a)\right)^2$.
- **Symboles** : $r$ = récompense ; $d\in\{0,1\}$ = 1 si l'épisode se termine ; $\gamma$ = facteur d'actualisation ; $Q_\theta$ = réseau en ligne (poids $\theta$) ; $Q_{\theta^-}$ = réseau cible (poids figés $\theta^-$) ; $s'$ = état suivant.
- **Paramètres** (appris) : poids $\theta$ du réseau. **Hyperparamètres** : dans votre code, seulement $\gamma=0{,}95$ ; un DQN complet en ajoute d'autres (taux d'apprentissage, taille de la mémoire et des lots, fréquence de copie du réseau cible, ε).
- **Avantage 1** : gère des espaces d'états grands ou continus (images, capteurs) grâce à la généralisation du réseau.
- **Avantage 2** : la mémoire de rejeu et le réseau cible stabilisent l'apprentissage.
- **Limite 1** : instable et gourmand en données et en calcul ; très sensible aux hyperparamètres.
- **Limite 2** : actions discrètes uniquement et tendance à surestimer les valeurs (d'où les variantes comme Double DQN).

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(11, 4.8))
ax.set_xlim(0, 11); ax.set_ylim(0, 6); ax.axis("off")

def boite(x, y, w, h, txt, c):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.1", fc=c, ec="black"))
    ax.text(x + w/2, y + h/2, txt, ha="center", va="center", fontsize=9)

boite(0.2, 2.2, 2.4, 1.6, "Mémoire de rejeu\n(s, a, r, s', fin)", "#dbe9f6")
boite(4.0, 4.0, 2.8, 1.4, "Réseau cible Q⁻\n(copie figée)", "#fde9c9")
boite(4.0, 0.6, 2.8, 1.4, "Réseau en ligne Q\n(entraîné)", "#fde9c9")
boite(7.6, 2.2, 3.1, 1.6, "Perte = (y − Q(s,a))²\ny = r + γ (1 − fin) · max Q⁻(s',·)", "#d7f0d2")
ax.annotate("", xy=(3.9, 4.6), xytext=(2.7, 3.6), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate("", xy=(3.9, 1.4), xytext=(2.7, 2.4), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate("", xy=(7.5, 3.5), xytext=(6.9, 4.6), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate("", xy=(7.5, 2.5), xytext=(6.9, 1.4), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(3.2, 5.0, "mini-lot", fontsize=8)
ax.text(5.4, 3.0, "copie périodique\nQ → Q⁻", ha="center", fontsize=8)
ax.annotate("", xy=(5.4, 3.9), xytext=(5.4, 2.1), arrowprops=dict(arrowstyle="->", ls="--", lw=1.2))
ax.text(9.15, 1.6, "gradient → met à jour Q", ha="center", fontsize=8)
ax.set_title("DQN : mémoire de rejeu, réseau en ligne et réseau cible (schéma conceptuel)")
plt.show()
```

**Lecture du schéma** : à gauche, la mémoire de rejeu stocke les transitions vécues ; on en tire un mini-lot. Le réseau cible (haut) fournit la valeur du meilleur coup suivant pour fabriquer la cible y ; le réseau en ligne (bas) fournit la valeur actuelle de l'action jouée. À droite, la perte mesure l'écart entre les deux, et son gradient met à jour le réseau en ligne. La flèche pointillée au centre signale que le réseau cible n'est qu'une copie du réseau en ligne, rafraîchie de temps en temps pour que la cible ne bouge pas à chaque pas.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique/gestion)** : piloter des décisions séquentielles à grand nombre d'états (gestion de stocks multi-produits, régulation de flux dans un entrepôt) à partir d'un simulateur.
- **Cas d'usage 2 (finance)** : stratégies d'exécution d'ordres ou d'allocation dynamique, en simulation, avec une grande prudence sur la fiabilité du simulateur.
- **Données et prérequis** : un simulateur ou un environnement interactif, beaucoup d'interactions, des états décrits par des vecteurs numériques ou des images, des actions discrètes ; un cadre de réseaux de neurones (PyTorch, TensorFlow).
- **Cas peu approprié** : petit espace d'états discret (Q-learning suffit), peu de données ou pas de simulateur, actions continues, ou contexte exigeant des décisions explicables.
- **Métriques adaptées** : récompense cumulée par épisode (moyenne glissante), perte de Bellman, stabilité sur plusieurs graines, évaluation sans exploration sur des épisodes séparés.

### Analyse de l'expérience Python
- **Entrées** : trois transitions fictives : récompenses [0, 1, −1], indicateurs de fin [Faux, Vrai, Faux], meilleures valeurs Q suivantes [2, 3, 1]. **Hyperparamètre** : `gamma=0.95`. Aucun réseau de neurones n'est entraîné (le commentaire du code le précise).
- **Sortie** : cibles de Bellman = [1,9 ; 1 ; −0,05], calculées par $r+\gamma(1-d)\max Q(s',\cdot)$.
- **Ce que cela permet** : vérifier le calcul à la main. Transition 1 : $0+0{,}95\times2=1{,}9$. Transition 2 : l'épisode se termine, donc le futur est ignoré et la cible vaut la récompense, 1 (la valeur 3 n'est pas utilisée). Transition 3 : $-1+0{,}95\times1=-0{,}05$, le futur compensant presque la pénalité.
- **Ce que cela ne permet pas** : aucune conclusion sur l'apprentissage ou la performance d'un DQN : les valeurs sont posées à la main, sans réseau, sans mémoire de rejeu ni réseau cible. Cela illustre seulement l'étape de calcul de la cible, dont dépend tout l'entraînement. Le choix de $\gamma$ n'est pas discuté.

  * **Évaluation Interne :** Mesure de la qualité des clusters sans étiquettes réelles en utilisant des métriques intrinsèques comme le *Score de Silhouette* ou l'index *Davies-Bouldin*.
  * **Évaluation Externe :** Si les étiquettes réelles (*ground truth*) sont disponibles, elles ne sont **jamais** utilisées lors du clustering, mais uniquement a posteriori pour évaluer la correspondance avec les métriques externes (*Adjusted Rand Index - ARI*, *NMI*).

* **Apprentissage par Renforcement (RL : Bandits, Q-learning, SARSA, DQN) :**

```python
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Chargement du jeu de données
california = fetch_california_housing(as_frame=True)
X, y = california.data, california.target

# 2. Découpage : 70% Train, 15% Validation, 15% Test
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42
)
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, test_size=0.1765, random_state=42
)  # 0.1765 * 0.85 ≈ 0.15

# 3. Standardisation sans fuite d'information
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# 4. Entraînement et choix des hyperparamètres sur Validation
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train_scaled, y_train)

val_preds = model.predict(X_val_scaled)
print(
    f"RMSE Validation : {np.sqrt(mean_squared_error(y_val, val_preds)):.4f}"
)

# 5. Évaluation finale sur l'ensemble de Test
test_preds = model.predict(X_test_scaled)
print(f"RMSE Test       : {np.sqrt(mean_squared_error(y_test, test_preds)):.4f}")
```

   * **K-Means vs DBSCAN :** K-Means impose des formes de clusters sphériques et nécessite de fixer $k$, tandis que DBSCAN détecte le bruit mais est sensible aux variations de densité.
   * **Apprentissage par Renforcement :** Forte sensibilité aux graines aléatoires et risque de forte variance d'un entraînement à l'autre.

```python
import time
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Affichage des versions des bibliothèques
import sklearn
from sklearn.cluster import DBSCAN, KMeans
from sklearn.datasets import fetch_california_housing, make_blobs
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, silhouette_score
from sklearn.model_selection import train_test_split
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

print("=== VERSIONS DES BIBLIOTHÈQUES ===")
print(f"NumPy      : {np.__version__}")
print(f"Pandas     : {pd.__version__}")
print(f"Scikit-Learn: {sklearn.__version__}\n")

SEED = 42

# ==============================================================================
# 1. RÉGRESSION : Référence (LinearRegression) vs Avancée (RandomForest)
# ==============================================================================
housing = fetch_california_housing(as_frame=True)
X, y = housing.data, housing.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Méthode de Référence : Linear Regression
t0 = time.time()
lr = LinearRegression()
lr.fit(X_train_scaled, y_train)
t_lr = time.time() - t0
y_pred_lr = lr.predict(X_test_scaled)

# Méthode Avancée : Random Forest Regressor
t0 = time.time()
rf = RandomForestRegressor(n_estimators=100, random_state=SEED, n_jobs=-1)
rf.fit(X_train_scaled, y_train)
t_rf = time.time() - t0
y_pred_rf = rf.predict(X_test_scaled)

# Tableau comparatif des métriques
reg_results = pd.DataFrame(
    {
        "Méthode": ["Linear Regression (Ref)", "Random Forest (Avancée)"],
        "MAE": [
            mean_absolute_error(y_test, y_pred_lr),
            mean_absolute_error(y_test, y_pred_rf),
        ],
        "RMSE": [
            np.sqrt(mean_squared_error(y_test, y_pred_lr)),
            np.sqrt(mean_squared_error(y_test, y_pred_rf)),
        ],
        "R²": [r2_score(y_test, y_pred_lr), r2_score(y_test, y_pred_rf)],
        "Temps Calcul (s)": [t_lr, t_rf],
    }
)

print("--- TABLEAU COMPARATIF : RÉGRESSION ---")
print(reg_results.to_string(index=False))

# Figure 1 : Comparaison Prédictions vs Réel
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(y_test, y_pred_lr, alpha=0.3, color="blue")
axes[0].plot([y.min(), y.max()], [y.min(), y.max()], "k--", lw=2)
axes[0].set_title("Régression Linéaire (Référence)")
axes[0].set_xlabel("Valeurs Réelles")
axes[0].set_ylabel("Prédictions")

axes[1].scatter(y_test, y_pred_rf, alpha=0.3, color="green")
axes[1].plot([y.min(), y.max()], [y.min(), y.max()], "k--", lw=2)
axes[1].set_title("Random Forest (Avancée)")
axes[1].set_xlabel("Valeurs Réelles")
axes[1].set_ylabel("Prédictions")

plt.suptitle(
    "Figure 1 : Comparaison des Prédictions - Régression Linéaire vs Random Forest",
    fontsize=14,
)
plt.tight_layout()
plt.show()

# ==============================================================================
# 2. RÉDUCTION DE DIMENSION ET CLUSTERING : ACP vs t-SNE / K-Means vs DBSCAN
# ==============================================================================
X_synth, _ = make_blobs(
    n_samples=500, centers=4, cluster_std=1.2, random_state=SEED
)

# Réduction de dimension
pca = PCA(n_components=2, random_state=SEED)
X_pca = pca.fit_transform(X_synth)
var_expl = np.sum(pca.explained_variance_ratio_)

tsne = TSNE(n_components=2, random_state=SEED)
X_tsne = tsne.fit_transform(X_synth)

# Clustering & Score de Silhouette
kmeans = KMeans(n_clusters=4, random_state=SEED, n_init=10)
labels_km = kmeans.fit_predict(X_synth)
sil_km = silhouette_score(X_synth, labels_km)

dbscan = DBSCAN(eps=0.8, min_samples=5)
labels_db = dbscan.fit_predict(X_synth)
# Silhouette uniquement s'il y a plus d'1 cluster (hors bruit)
sil_db = (
    silhouette_score(X_synth, labels_db)
    if len(set(labels_db)) > 1
    else float("nan")
)

print("\n--- MÉTRIQUES CLUSTERING & RÉDUCTION DE DIMENSION ---")
print(f"ACP - Variance Expliquée Cumulée (2 comp) : {var_expl*100:.2f}%")
print(f"K-Means - Score de Silhouette               : {sil_km:.4f}")
print(f"DBSCAN  - Score de Silhouette               : {sil_db:.4f}")

# Figure 2 : Visualisations
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=labels_km, cmap="viridis", s=30)
axes[0].set_title(
    f"ACP + K-Means (Silhouette: {sil_km:.2f} | Var Expl: {var_expl*100:.1f}%)"
)

axes[1].scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels_db, cmap="plasma", s=30)
axes[1].set_title(
    f"t-SNE + DBSCAN (Conservation Voisinages / Détection Bruit)"
)

plt.suptitle(
    "Figure 2 : Comparaison Réduction de Dimension (ACP vs t-SNE) & Clustering (K-Means vs DBSCAN)",
    fontsize=13,
)
plt.tight_layout()
plt.show()
```
